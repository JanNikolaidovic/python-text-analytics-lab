### Numeric Types: `int` and `float`

Python splits numbers into two simple data types, depending on whether they have a decimal point:

| Type | Example | Typically Used For |
|---|---|---|
| `int` (integer) | `150` | Whole quantities — shares held, number of employees, units sold |
| `float` (floating-point) | `42.75` | Values that need decimals — prices, percentages, exchange rates |

Just like with strings, Python figures out which type you mean automatically, based on how you write the value — no decimal point means `int`, any decimal point at all means `float`. Let's check this with the `type()` function you met in Lab 01. Run the code cell below:

```python
shares_held = 150
share_price = 42.75

print(type(shares_held))
print(type(share_price))
```

One small detail that trips people up: a value is a `float` the moment it's *written* with a decimal point, even if that decimal happens to be zero. `42.0` is a `float`, while `42` is an `int` — same number, different type. Keep an eye out for this, since it explains a couple of surprises we'll run into shortly.

### Arithmetic Operators

Python supports all the basic math operations you'd expect, plus a couple of business-handy extras:

| Operator | Name | What It Does | Example | Result |
|---|---|---|---|---|
| `+` | Addition | Adds two numbers | `10 + 3` | `13` |
| `-` | Subtraction | Subtracts the right number from the left | `10 - 3` | `7` |
| `*` | Multiplication | Multiplies two numbers | `10 * 3` | `30` |
| `/` | Division | Divides the left number by the right | `10 / 3` | `3.333...` |
| `//` | Floor Division | Divides, then rounds *down* to the nearest whole number | `10 // 3` | `3` |
| `%` | Modulo | Gives the *remainder* left over after floor division | `10 % 3` | `1` |
| `**` | Exponent | Raises the left number to the power of the right | `10 ** 3` | `1000` |

Let's put the four most familiar ones to work on a simple quarterly report. Run the code cell below:

```python
revenue = 12500
costs = 9800

profit = revenue - costs
combined_two_quarters = revenue + revenue
average_monthly_revenue = revenue / 3

print(profit)
print(combined_two_quarters)
print(average_monthly_revenue)
```

Notice that `average_monthly_revenue` came out as `4166.666666666667`, a `float` — even though both `revenue` and `3` are whole numbers. In Python, the `/` operator **always** produces a `float`, no matter what you feed it, precisely so you never accidentally lose the decimal part of a division. If you need the two original integers to combine into another integer, `+`, `-`, and `*` will do that; only `/` promotes the result to a `float` automatically.

### Splitting Things Evenly: Floor Division and the Remainder

Sometimes plain division isn't what you actually want. Imagine you need to split a bonus pool evenly among a team, in whole currency units, and figure out how much is left over afterward. That's exactly what `//` and `%` are for:

* **`//` (floor division)** divides and then rounds *down* to the nearest whole number — how many whole times the amount fits.
* **`%` (modulo)** gives you the *remainder* — whatever is left over after that even split.

Run the code cell below to split a bonus pool among a team:

```python
bonus_pool = 10000
team_size = 7

per_employee = bonus_pool // team_size
leftover = bonus_pool % team_size

print(per_employee)
print(leftover)
```

Each of the 7 team members gets `1428`, and `4` currency units are left in the pool — `1428 * 7 + 4` does indeed add back up to `10000`. This combination of `//` and `%` shows up constantly whenever you need to distribute something evenly and account for the remainder, and `%` alone is also the standard trick for checking whether a number is even or odd (`number % 2` gives `0` for even numbers and `1` for odd ones).

### Exponents: `**`

The last operator in our table, `**`, raises a number to a power — useful for anything that compounds, like interest. Run the code cell below to project an investment's value after two years of 5% annual growth:

```python
principal = 1000
annual_growth = 1.05

balance_after_two_years = principal * annual_growth ** 2
print(balance_after_two_years)
```

### Order of Operations

That last calculation worked correctly without any parentheses: `principal * annual_growth ** 2` computed `annual_growth ** 2` *first*, and only then multiplied the result by `principal`. This isn't an accident — Python follows the same order of operations you learned in math class (often remembered as PEMDAS): exponents (`**`) before multiplication and division (`*`, `/`, `//`, `%`), which happen before addition and subtraction (`+`, `-`).

Whenever you want to force a different order, wrap the part that should happen first in parentheses `()`. Let's see this make an actual difference. Run the code cell below:

```python
subtotal = 200
discount = 50

print(subtotal - discount * 0.5)
print((subtotal - discount) * 0.5)
```

Two nearly identical lines, two very different answers. In the first, Python computes `discount * 0.5` (`25`) *before* subtracting, giving `175`. In the second, the parentheses force the subtraction to happen first (`150`), and *that* result gets halved, giving `75`. When in doubt about how an expression will be evaluated, adding parentheses costs you nothing and removes all ambiguity — both for Python and for whoever reads your code later.

### Comparing Values: Comparison Operators

So far, every operator we've used has *combined* numbers into a new number. **Comparison operators** do something different: they ask a yes-or-no question about two values.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Is equal to | `10 == 10` | `True` |
| `!=` | Is not equal to | `10 != 5` | `True` |
| `<` | Is less than | `10 < 5` | `False` |
| `>` | Is greater than | `10 > 5` | `True` |
| `<=` | Is less than or equal to | `10 <= 10` | `True` |
| `>=` | Is greater than or equal to | `10 >= 15` | `False` |

Note the double equals sign, `==`, for checking equality — a single `=` is reserved entirely for assignment (giving a variable a value), so Python uses `==` to ask "are these the same?" without any ambiguity.

Let's check whether a company hit its quarterly target. Run the code cell below:

```python
quarterly_revenue = 125000
target = 120000

print(quarterly_revenue > target)
print(quarterly_revenue == target)
```

`True` and `False` are the only two possible values of Python's third simple data type: the **Boolean** (`bool`, named after the mathematician George Boole, which I encourage you to Google and learn about). Every comparison you just ran produces a Boolean, and you can also create one directly, without any comparison at all, just by writing `True` or `False` as a value. Run the code cell below:

```python
is_profitable = True
missed_deadline = False

print(is_profitable)
print(type(is_profitable))
print(type(quarterly_revenue > target))
```

Both routes — writing `True`/`False` directly, or letting a comparison produce one — give you the exact same `bool` type. Booleans might look simple, but they're the foundation for **conditional logic**: code that makes decisions and does different things depending on whether something is `True` or `False`. You'll meet Python's `if` statements (and the `and`, `or`, and `not` keywords for combining multiple Booleans together) in a future lab — for now, just get comfortable recognizing a Boolean when you see one, and producing one with a comparison.

### Mixing Numbers and Text

You'll very often want to report a number as part of a sentence — "Total revenue: 125000" reads a lot better than a number sitting alone on its own line. It's natural to reach for the `+` concatenation you learned in Part 1, the same way you joined two strings together. Let's see what happens. Run the code cell below and read the error carefully:

```python
total_revenue = 125000
print("Total revenue: " + total_revenue)
```

Python stopped you with a `TypeError: can only concatenate str (not "int") to str`. Unlike `.title()` or `.upper()`, which only ever hand you back another string, `+` between a string and a number genuinely confuses Python — should `"Total revenue: " + 125000` mean text-plus-number, or is it trying to *add* two things that aren't compatible? Rather than guess, Python simply refuses.

You already know the fix: **f-strings** don't have this problem at all, because you're not concatenating anything — you're just dropping a value into a slot inside your text, and Python converts it to a readable form automatically, whatever type it is.

```python
print(f"Total revenue: {total_revenue}")
```

### Rounding Numbers: The `round()` Function

Division and averages have a habit of producing long, awkward decimals — `4166.666666666667` isn't something you'd want to show a client. Python offers `round(value, digits)` for exactly this: a general-purpose **function**, not a method (remember that distinction from Part 1?), that rounds a number to however many decimal places you specify.

There's also a shortcut for doing this purely for *display*, right inside an f-string: adding `:.2f` after the variable name formats it with exactly 2 decimal places. Let's compare both. Run the code cell below:

```python
average_price = 19.98765

rounded_price = round(average_price, 2)
print(rounded_price)
print(type(rounded_price))

print(f"${average_price:.2f}")
```

Both approaches display `19.99`, but they're doing different jobs. `round()` creates an actual new number — `rounded_price` is still a `float` you can keep calculating with. The `:.2f` inside the f-string, on the other hand, only changes how `average_price` is *displayed* in that one `print()`; the original variable is completely untouched, still holding all those extra decimals. Use `round()` when the rounded value itself matters for later math, and `:.2f` when you just need a clean number to show someone.
