### Part 1: Arrays

NumPy is always imported under the short name `np` (the `import ... as` pattern from Lab 06). You create an array by passing a list to `np.array()`:

```python
import numpy as np

prices = np.array([148.0, 38.0, 198.0, 98.0, 68.0])

print(prices)
print(type(prices))
```

It prints like a list without the commas. Two attributes describe every array:

* **`.shape`**: its size along each dimension, as a tuple. A 1-D array of 5 numbers has shape `(5,)`.
* **`.dtype`**: the type of **every** element. Unlike a list, an array holds one type only, which is part of what makes it fast.

```python
print(prices.shape)
print(prices.dtype)
print(len(prices))

ratings = np.array([5, 4, 5, 1, 3])
print(ratings.dtype)
```

`float64` is a decimal number and `int64` a whole number (the `64` is how many bits each one uses in memory; you can ignore it). If you mix integers and floats in one array, NumPy converts everything to float so that the type stays uniform.

### Vectorized Arithmetic

Here's the big difference from lists. Arithmetic on an array is applied to **every element**:

```python
price_list = [148.0, 38.0, 198.0]
price_array = np.array([148.0, 38.0, 198.0])

print(price_list * 2)
print(price_array * 2)
```

`*` on a list **repeats** the list; `*` on an array **multiplies each number**. The same goes for `+`: two lists get glued together, while two arrays of the same shape are added position by position.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print([1, 2, 3] + [10, 20, 30])
print(a + b)
print(a * b)
print(b / a)
print(a ** 2)
```

No loops, no comprehensions. Doing an operation on whole arrays at once is called **vectorization**, and it's the single most important idea in NumPy (and in pandas, which inherits it).

A realistic example: add 24% VAT to every price, and round to 2 decimals.

```python
prices_with_vat = np.round(prices * 1.24, 2)
print(prices_with_vat)
```

Comparisons are vectorized too. The result is an array of Booleans, one for each element:

```python
print(prices > 100)
```

An array of `True`/`False` values like this is called a **mask**, and it's the key to the next part.

### Part 2: Indexing, Slicing, and Boolean Masks

Indexing and slicing work exactly as they do for lists (Lab 04):

```python
prices = np.array([148.0, 38.0, 198.0, 98.0, 68.0])

print(prices[0])
print(prices[-1])
print(prices[1:4])
```

### Boolean Masks

Now the new part. Put a mask (an array of Booleans, same length) inside the square brackets, and you get back **only the elements where the mask is `True`**:

```python
expensive = prices > 100

print(expensive)
print(prices[expensive])
print(prices[prices < 50])
```

Read `prices[prices < 50]` as *"prices, where the price is below 50"*. This filtering is something you'll do constantly in pandas, with exactly the same syntax.

To combine conditions, use **`&`** (and) and **`|`** (or), with **parentheses around each condition**. The words `and` and `or` don't work on arrays.

```python
print(prices[(prices > 50) & (prices < 150)])
print(prices[(prices < 50) | (prices > 150)])
```

> **Forgetting the parentheses is the most common NumPy (and pandas) mistake.** `prices > 50 & prices < 150` is evaluated in the wrong order and raises a confusing error. Always write `(condition) & (condition)`.

### Counting With Masks

Because `True` counts as `1` and `False` as `0` (you used this in Lab 09), two very useful tricks follow:

* `mask.sum()` is **how many** elements satisfy the condition,
* `mask.mean()` is **what share** of the elements satisfy it.

```python
print((prices > 100).sum())
print((prices > 100).mean())
```

Two out of five prices, or `0.4` (40%), are above 100.

### Part 3: Summarizing Data With Aggregations

An **aggregation** reduces a whole array to one number: its sum, its average, its largest value. NumPy has all the standard ones, either as methods (`arr.mean()`) or as functions (`np.median(arr)`).

Let's use real numbers. Load the 600 reviews from Labs 08 and 09 again, and make two arrays: the star `ratings`, and the `lengths` of the reviews in words.

```python
import urllib.request

DATA_URL = "https://raw.githubusercontent.com/JanNikolaidovic/python-text-analytics-lab/main/data/reviews_sample.txt"
urllib.request.urlretrieve(DATA_URL, "reviews_sample.txt")

rating_list = []
length_list = []
with open("reviews_sample.txt", encoding="utf-8") as file_object:
    for line in file_object:
        rating, text = line.rstrip("\n").split("\t")
        rating_list.append(int(rating))
        length_list.append(len(text.split()))

ratings = np.array(rating_list)
lengths = np.array(length_list)
print(ratings.shape, lengths.shape)
print(ratings[:10])
print(lengths[:10])
```

Now the summaries. Each one is a single call:

```python
print(f"Mean length:    {lengths.mean():.1f} words")
print(f"Median length:  {np.median(lengths):.1f} words")
print(f"Std deviation:  {lengths.std():.1f} words")
print(f"Shortest:       {lengths.min()} words")
print(f"Longest:        {lengths.max()} words")
print(f"Total words:    {lengths.sum()}")
```

A quick guide to reading these:

* The **mean** is the familiar average. The **median** is the middle value when everything is sorted: half the reviews are shorter, half are longer. When the two are close (as here), the data is roughly balanced around the middle. When the mean is much larger than the median, a few very large values are pulling it up.
* The **standard deviation** (`std`) measures how spread out the values are around the mean, in the same units (words). Roughly, a "typical" review is within about one standard deviation of the mean.
* Notice that the longest review is only 110 words. For real reviews that's suspicious: surely *someone* wrote an essay? In fact, the store's website limited reviews to about 500 characters, so long reviews were cut off. Noticing things like this is exactly what exploratory data analysis is about.

### Where Is the Maximum? `argmax()`

`.max()` tells you the largest value; **`.argmax()`** tells you its **position**. That's what you need to go back and look at the record itself:

```python
position = lengths.argmax()
print(position)
print(lengths[position], ratings[position])
```

### Counting Categories: `np.unique()`

For a column of categories (like star ratings), `np.unique(..., return_counts=True)` returns the distinct values and how often each appears, as two arrays. It's NumPy's version of `Counter`:

```python
values, counts = np.unique(ratings, return_counts=True)
print(values)
print(counts)
print(np.round(counts / counts.sum(), 3))
```

Dividing the counts by their total turns them into **shares** that add up to 1. Both lines are arrays of the same shape, so the division happens element by element.

### Comparing Groups With Masks

Masks and aggregations together answer comparison questions. Do unhappy customers write longer reviews?

```python
for star in range(1, 6):
    group_lengths = lengths[ratings == star]
    print(f"{star} stars: {len(group_lengths):>3} reviews, mean length {group_lengths.mean():.1f} words")
```

`lengths[ratings == star]` uses a mask built from **one** array to select from **another**. That works because both arrays are in the same order: position `i` in each describes the same review. This "select one column using a condition on another" move is the heart of data analysis, and pandas will make it even easier in the next lab.

Look at the numbers: the differences are small. Every group averages somewhere around 58 to 65 words, with 5-star reviews the shortest and 4-star reviews the longest. With only 20 one-star reviews in this sample, a difference of a few words could easily be chance. That's a useful habit to build: **before you explain a difference, ask whether it's big enough to need explaining.** You'll look at this again on all ~23,000 reviews later.

### Part 4: Vectors in NumPy

Everything you wrote by hand in Lab 10 is built into NumPy:

| Lab 10 (plain Python) | NumPy |
|---|---|
| `dot(a, b)` | `np.dot(a, b)`, or `a @ b` |
| `norm(a)` | `np.linalg.norm(a)` |
| `cosine(a, b)` | `a @ b / (np.linalg.norm(a) * np.linalg.norm(b))` |

The `@` operator means "dot product" (for 2-D arrays it's matrix multiplication, which you'll use in Part 5). `linalg` stands for **linear algebra**, the branch of maths that deals with vectors and matrices.

```python
a = np.array([3, 0, 1])
b = np.array([2, 5, 1])

print(np.dot(a, b))
print(a @ b)
print(np.linalg.norm(np.array([3, 4])))
```

Same results as Lab 10 (`7` and `5.0`). Cosine similarity is now one line:

```python
def cosine(a, b):
    return a @ b / (np.linalg.norm(a) * np.linalg.norm(b))

print(round(cosine(np.array([3, 0]), np.array([1, 1])), 2))
print(round(cosine(np.array([1, 1, 0]), np.array([4, 4, 0])), 2))
```

`0.71` for the 45° angle and `1.0` for two vectors with the same direction, exactly as in Lab 10.

### Part 5: 2-D Arrays, a Document-Term Matrix

A 2-D array is a grid of numbers with **rows and columns**, like a spreadsheet. Pass `np.array()` a list of lists, and each inner list becomes a row. The document-term matrix from Lab 10 (one row per review, one column per word) is exactly this.

The cell below rebuilds the eight real reviews from Lab 10 as a document-term matrix `X`, with stopwords removed. The first part is the same Lab 10 code, condensed; the last line turns the list of vectors into one 2-D array.

```python
import re

reviews = [
    "Gorgeous top but runs large, especially in the arm holes.",
    "This is an attractive shirt but runs big. i ordered the black and white.",
    "This dress is really really soft and comfortable. it's loose but still flattering. i would definitely recommend it.",
    "Love this sweater! great for work. very soft and comfortable. color is gorgeous!",
    "This shirt is unflattering and boxy. it does not look cute at all on me. i am returning.",
    "I'm returning this cardigan. it doesn't fit me at all. it's shapeless and loose.",
    "Great casual weekend pants and super comfy and great color.",
    "The print on these leggings is so cool but they are completely see-through. i mean seriously, seriously see-through.",
]
STOPWORDS = {"a", "an", "the", "and", "but", "is", "it", "it's", "this", "i", "i'm", "me",
             "on", "in", "for", "so", "at", "all", "they", "these", "am", "are", "very", "would"}

token_lists = [[t for t in re.findall(r"[a-z']+", text.lower()) if t not in STOPWORDS] for text in reviews]
vocab = sorted(set(t for tokens in token_lists for t in tokens))

X = np.array([[tokens.count(word) for word in vocab] for tokens in token_lists])
print(X.shape)
```

The shape `(8, 51)` reads as **(rows, columns)**: 8 reviews, 51 vocabulary words.

Indexing a 2-D array takes two positions, `X[row, column]`, and a `:` on its own means "everything along this dimension":

```python
print(X[0])
print(X[0, :5])

great = vocab.index("great")
print(X[:, great])
```

* `X[0]` is the whole first row: review 0's vector.
* `X[0, :5]` is row 0, first five columns.
* `X[:, great]` is **every row** in the column for the word `great`: how many times each review uses it. (`vocab.index("great")` finds the column number, using the list's `.index()` method.)

### Aggregating Along an Axis

On a 2-D array, an aggregation can work on the whole grid, or **along one axis**:

* `axis=0` collapses the **rows**, giving one result **per column** (here: per word),
* `axis=1` collapses the **columns**, giving one result **per row** (here: per review).

```python
word_totals = X.sum(axis=0)
review_lengths = X.sum(axis=1)

print(word_totals.shape, review_lengths.shape)
print(review_lengths)
```

`word_totals` has one number per word (51) and `review_lengths` one per review (8). To find the most-used words, sort the totals. **`np.argsort()`** returns the **positions** that would sort an array (smallest first), so `[::-1]` reverses them to get the largest first:

```python
top_positions = np.argsort(word_totals)[::-1][:5]

for position in top_positions:
    print(vocab[position], word_totals[position])
```

### All Pairwise Similarities at Once

In Lab 10 you compared every pair of reviews with two nested loops. With NumPy, you can compute **all** pairs in two steps:

1. Turn every row into a **unit vector** by dividing it by its norm.
2. Multiply the matrix by its own **transpose** (`X.T`, the matrix flipped so rows become columns). Every entry of the result is the dot product of one pair of unit vectors, which is exactly their cosine similarity.

```python
row_norms = np.linalg.norm(X, axis=1, keepdims=True)
X_unit = X / row_norms

similarity = X_unit @ X_unit.T

print(row_norms.shape)
print(similarity.shape)
print(np.round(similarity, 2))
```

Some details worth understanding:

* `np.linalg.norm(X, axis=1)` computes one norm per row. `keepdims=True` keeps the result as a column of shape `(8, 1)` rather than a flat `(8,)`.
* `X / row_norms` divides an `(8, 51)` array by an `(8, 1)` array. NumPy "stretches" the single column across all 51 columns so that every entry of row `i` is divided by row `i`'s norm. This stretching is called **broadcasting**; it's the same thing that lets you write `prices * 1.24`, where a single number is stretched across the whole array.
* `similarity` is an `8 × 8` table: row `i`, column `j` is the cosine similarity between review `i` and review `j`. The diagonal is all `1`s (every review is identical to itself), and the table is symmetric (`similarity[i, j]` equals `similarity[j, i]`).

To find the most similar **pair**, ignore the diagonal (set it to 0) and look for the largest entry. `argmax()` on a 2-D array returns a position in the flattened grid; `np.unravel_index()` turns it back into a `(row, column)` pair:

```python
np.fill_diagonal(similarity, 0)

i, j = np.unravel_index(similarity.argmax(), similarity.shape)
print(f"Most similar ({similarity[i, j]:.2f}):")
print("  ", reviews[i])
print("  ", reviews[j])
```

The same pair and the same score (`0.34`) as the stopword version in Lab 10, but with no loops at all. With 8 reviews the difference hardly matters; with 20,000 reviews, the loop version would take far longer, as the next part shows.

### Part 6: Why Bother? Speed

Vectorized code is shorter, but the main reason NumPy exists is **speed**. Its loops run in compiled C code instead of Python. Let's time the sum of squares of one million numbers both ways, using `time.perf_counter()` (a stopwatch from the standard library).

```python
import time

numbers = list(range(1_000_000))
numbers_array = np.array(numbers)

start = time.perf_counter()
total = 0
for n in numbers:
    total += n * n
python_seconds = time.perf_counter() - start

start = time.perf_counter()
total_np = (numbers_array * numbers_array).sum()
numpy_seconds = time.perf_counter() - start

print(total == total_np)
print(f"Python loop: {python_seconds:.4f} s")
print(f"NumPy:       {numpy_seconds:.4f} s")
print(f"NumPy was about {python_seconds / numpy_seconds:.0f} times faster.")
```

The exact numbers depend on the computer and the operation, but NumPy is usually somewhere between several and a hundred times faster, and the gap grows with more complex calculations. That's the difference between an analysis that runs in a second and one that runs over a coffee break, and it's why pandas, scikit-learn, and nearly every other data library are built on top of NumPy arrays.

### Recap: Your NumPy Toolkit

| Tool | What It Does | Example |
|---|---|---|
| `np.array(list)` | Create an array (a list of lists gives 2-D) | `np.array([1, 2, 3])` |
| `.shape` / `.dtype` | Size along each dimension / element type | `X.shape` → `(8, 51)` |
| Vectorized arithmetic | `+ - * / **` apply to every element | `prices * 1.24` |
| Comparison → mask | Boolean array, one value per element | `prices > 100` |
| `arr[mask]` | Keep only the elements where the mask is `True` | `prices[prices > 100]` |
| `&` / `\|` | Combine masks (always use parentheses) | `(a > 1) & (a < 5)` |
| `mask.sum()` / `mask.mean()` | How many / what share are `True` | `(ratings == 5).mean()` |
| `.mean()` `.std()` `.min()` `.max()` `.sum()` | Aggregations | `lengths.mean()` |
| `np.median(arr)` | Middle value | `np.median(lengths)` |
| `.argmax()` / `np.argsort()` | Position of the max / positions that sort | `np.argsort(x)[::-1]` |
| `np.unique(arr, return_counts=True)` | Distinct values and their counts | `values, counts = ...` |
| `a @ b` / `np.dot(a, b)` | Dot product (matrix product for 2-D) | `X_unit @ X_unit.T` |
| `np.linalg.norm(a)` | Vector norm (`axis=1` for one per row) | `np.linalg.norm(X, axis=1)` |
| `X[i, j]`, `X[i]`, `X[:, j]` | One entry / one row / one column | `X[:, great]` |
| `axis=0` / `axis=1` | Aggregate down columns / across rows | `X.sum(axis=0)` |
| `.T` | Transpose (rows become columns) | `X.T` |
| Broadcasting | Stretch a smaller array to match a larger one | `X / row_norms` |

The whole lab in one sentence: **a NumPy array lets you do arithmetic, comparisons, filtering, and summaries on every value at once, which turns Lab 10's loops into single lines and makes them fast enough for real datasets.**
