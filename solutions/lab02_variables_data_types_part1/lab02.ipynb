### **Strings and Text**

Almost every computer program exists to take in some kind of data, process it, and produce a useful result. Because data comes in so many different forms, programming languages classify it into specific **data types**.

The very first data type we will explore is the **string**. While strings might seem basic at first glance, they are incredibly versatile and you will use them constantly!

### What is a String?
At its core, a string is simply a sequence of characters (which can include letters, numbers, punctuation, or even spaces). In Python, absolutely anything wrapped inside quotation marks is instantly recognized as a string. 

Python is wonderfully flexible and allows you to use either double quotes (`" "`) or single quotes (`' '`) to define your strings. For example:
* `"Welcome to the data analytics lab."`
* `'Learning to code is a rewarding journey.'`

### The Power of Flexible Quotes
Why does Python give you two different ways to do the exact same thing? This clever flexibility allows you to easily include actual quotes or apostrophes *inside* your text without breaking your code! 

If your text contains a single quote (like an apostrophe), you simply wrap the entire string in double quotes. If you need to quote someone using double quotes, you wrap the string in single quotes. 

Here is how that works in practice:
* `'The manager said, "Please send me the report by Friday!"'`
* `"I'm really enjoying this chapter on data types."`
* `"The word 'algorithm' sounds intimidating, but it is just a set of instructions."`

### Changing Case in a String Using Methods

One of the most common (and easiest!) tasks when cleaning real-world business data is standardizing the capitalization of your text. 

Let's imagine you scraped some financial news headlines from a website, but the formatting is completely completely flat and lowercase. We can fix this using a Python **method**. A method is simply a specific action that Python can perform on a piece of data.

Run the code cell below to see how we can instantly format a messy news headline:

```python
news_headline = "central bank leaves interest rates unchanged"
print(news_headline.title())
```

If you ran the cell above, you should see the headline printed beautifully as:
`Central Bank Leaves Interest Rates Unchanged`

So, what exactly happened behind the scenes? Inside the `print()` command, we attached `.title()` to our variable. Let's break down that syntax:
* **The Dot (`.`):** The dot directly connects the variable to the method, telling Python to make the `title()` action act upon our `news_headline` string.
* **The Parentheses `()`:** Every method is followed by a set of parentheses. Often, methods need extra information to do their job, which you would place inside these parentheses. The `title()` method, however, knows exactly what to do (capitalize the first letter of every word) without any extra instructions, so its parentheses are left perfectly empty!

### The Upper and Lower Methods

Python offers several other handy methods for adjusting text case, which are absolute lifesavers in data analytics. Two of the most heavily used are `upper()` and `lower()`. 

* **The `upper()` method** is perfect for standardizing financial identifiers. For example, if a user types a stock ticker in lowercase, you can instantly convert it to the standardized uppercase format (like turning "aapl" into "AAPL").
* **The `lower()` method** is an absolute powerhouse for natural language processing. If you are analyzing thousands of customer reviews to gauge product sentiment, you can never trust the users' random capitalization. A common industry best practice is to convert all incoming review text to lowercase before analyzing it, ensuring that "OUTSTANDING!" and "outstanding!" are treated as the exact same word by your algorithm.

Let's see both of these in action! Run the code cell below:

```python
stock_ticker = "tsla"
print(stock_ticker.upper())

customer_review = "OUTSTANDING QUARTERLY EARNINGS!"
print(customer_review.lower())
```

### Combining or Concatenating Strings

When working with datasets, you will frequently need to combine multiple pieces of textual data. For example, imagine you pull a list of customers from a database where their first and last names are stored in completely separate columns, but you need to combine them to generate a personalized email. 

In Python, you can merge strings together using the plus symbol (`+`). This action is formally called **concatenation**.

Let's see how we can use the `+` symbol to combine a first name, a space, and a last name into a single variable. Run the code cell below:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

print(full_name)
```

If you ran the code, you should see the output `jane smith`. 

Concatenation is incredibly powerful when you want to automate complete messages, like a customized greeting for a marketing campaign. We can even combine concatenation with the string methods we learned earlier (like `.title()`) to ensure the final output is perfectly formatted!

Let's build a dynamic greeting. Run the following code block:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

print("Dear " + full_name.title() + ", welcome to our premium financial services!")
```

While the code above works perfectly and returns a beautifully formatted greeting (`Dear Jane Smith...`), writing long chains of text directly inside the `print()` function can get messy and hard to read. 

A much cleaner, professional approach is to build your entire message using concatenation, store that final message inside a new variable, and *then* print it. This makes your code much easier to manage.

Let's look at how we can store an automated email message into a single variable:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

automated_email = "Dear " + full_name.title() + ", your monthly portfolio report is attached."
print(automated_email)
```

### A Cleaner Approach: Introduction to f-strings

While combining strings with the `+` symbol works perfectly fine, it can quickly become tedious. Manually keeping track of quotation marks and making sure you didn't forget a space between words can lead to messy, error-prone code.

Fortunately, modern Python offers a much more elegant and readable solution: **f-strings** (short for *formatted string literals*). 

To create an f-string, you simply place the letter `f` immediately before your opening quotation mark. Then, you can inject your variables directly inside the text by wrapping them in curly braces `{}`. Python will automatically replace those braces with the variable's actual value!

Let's rewrite our automated email from the previous example using this cleaner approach. Run the code cell below:

```python
first_name = "jane"
last_name = "smith"

#combining names smoothly with an f-string
full_name = f"{first_name} {last_name}"

#building the final message directly
automated_email = f"Dear {full_name.title()}, your monthly portfolio report is attached."

print(automated_email)
```

Notice how much easier that was to read! Instead of stitching together separate pieces with `+`, we wrote the sentence almost exactly as we wanted it to appear, and simply dropped `{first_name.title()}` and `{last_name.title()}` in the spots where we needed the actual values. You can even call methods, like `.title()`, directly *inside* the curly braces.

From this point on in the course, we will almost always reach for f-strings instead of `+` concatenation — get comfortable with them now, because you will be using them constantly.

### Special Characters and Multi-line Text

Sometimes the text you are working with needs to contain characters that would otherwise confuse Python — like a line break, a tab, or a quotation mark that matches the ones wrapping your string. Python handles this with **escape sequences**: a backslash `\` followed by a character, which together mean something special rather than being printed literally.

Two of the most useful are:
* `\n` — inserts a new line
* `\t` — inserts a tab (handy for lining up columns of output)

For genuinely long stretches of text that span **multiple lines** — think of the body of an email, or a full customer review — retyping `\n` over and over gets tedious. Python gives you a more comfortable option: wrap the text in **triple quotes** (`"""` or `'''`), and you can simply press Enter and keep typing across as many lines as you like, exactly as the text should appear.

Let's see both ideas in action. Run the code cell below:

```python
report_line = "Q3 Revenue:\t$4.2M\nQ3 Net Income:\t$0.8M"
print(report_line)

email_body = """Dear Jane Smith,

Thank you for being a valued client. Your Q3 portfolio report is attached.

Best regards,
The Investment Team"""
print(email_body)
```

### How Long Is This Text? The `len()` Function

Businesses frequently need to measure the *length* of a piece of text — think of a social media post that must fit within a strict character limit, or an account number that must contain exactly the right number of digits to be considered valid.

Python's built-in `len()` function counts exactly how many characters are inside a string (letters, numbers, spaces, and punctuation all count). Unlike the methods we've used so far, `len()` is *not* attached to the string with a dot — instead, you pass the string to it inside the parentheses.

Let's check whether a draft social media post fits inside a 100-character limit. Run the code cell below:

```python
post = "Our Q3 earnings beat analyst expectations, driven by strong growth in our core markets."
print(len(post))
print(len(post) <= 100)
```

### Why Isn't `len()` a Method?

You might have noticed that `len(post)` doesn't look like the other tools you've used so far, such as `post.upper()`. That's not an accident. The distinction comes down to *what* each tool needs to do its job.

Methods like `.upper()`, `.strip()`, and `.replace()` are attached to a specific data type, because they only make sense for that type — `.title()` case only means something for text. `len()`, on the other hand, is a general-purpose **function** that works the exact same way across many different data types: it counts characters in a string today, and later in the course it will count items in a list or rows in a dataset. Rather than reimplementing that same job as a separate method on every single type, Python provides it once as a standalone function that any of them can plug into.

A simple rule of thumb: if it's written as `object.something()`, it is a **method** belonging specifically to that object's type. If it's written as `something(object)`, it's a **function** Python offers more broadly.

You'll notice the second line printed `True` rather than a number or a piece of text. That's Python's third simple data type in action: the **Boolean**, which can only ever be `True` or `False`. We'll dig into Booleans properly in Part 2, once we've covered the comparison operators (like `<=`) that produce them.

### Cleaning Extra Whitespace and Fixing Typos: `strip()` and `replace()`

Real-world text data is rarely as tidy as the examples we've used so far. Survey responses, spreadsheet exports, and web-scraped text are notorious for carrying invisible extra spaces at the beginning or end, and for containing exact phrases you need to correct or standardize before you can trust the data.

Two more string methods handle these situations beautifully:

* **`strip()`** removes any whitespace (spaces, tabs, or line breaks) from the very beginning and end of a string — but leaves spaces *inside* the text untouched.
* **`replace(old, new)`** searches through a string and swaps every occurrence of `old` with `new`. Unlike the methods we've seen so far, `replace()` needs two pieces of information inside its parentheses: the exact text to find, and the text to replace it with.

One more handy trick: since every one of these methods returns a brand-new string, you can **chain** them together with extra dots, all in a single line — for example `raw_text.strip().lower()` cleans up whitespace and lowercases the text in one step.

Let's clean up a messy survey response that was exported with stray whitespace, and correct a company name that was consistently misspelled. Run the code cell below:

```python
raw_response = "   The custmer servide team was extremely helpful!   "
cleaned = raw_response.strip()
corrected = cleaned.replace("custmer servide", "customer service")

print(corrected)
```

### One-Sided Trimming: `lstrip()` and `rstrip()`

Sometimes you only want to trim whitespace from one side of a string. `.lstrip()` removes whitespace from the **left** (the start) only, while `.rstrip()` removes it from the **right** (the end) only — the `.strip()` you just used is really just doing both of these at once.

Let's compare all three side by side on the same padded value. We'll wrap each result in square brackets so any leftover whitespace is easy to spot. Run the code cell below:

```python
padded = "   $4,200.00   "

print(f"[{padded.lstrip()}]")
print(f"[{padded.rstrip()}]")
print(f"[{padded.strip()}]")
```

### Checking Whether a Word Appears: the `in` Keyword

Often, you don't need to extract or transform text — you just need a quick yes-or-no answer to whether a particular word or phrase shows up somewhere inside it. Does this customer review mention the word "refund"? Does this news headline mention a specific company?

Python's `in` keyword checks exactly that. It compares a smaller string against a larger one and returns `True` if the smaller string appears anywhere inside the larger one, or `False` if it doesn't. This check is **case-sensitive**, which is exactly why the `.lower()` method you learned earlier is so often used right before it.

Let's check a customer review for a word that might need urgent attention. Run the code cell below:

```python
review = "I love the product, but the refund process was incredibly slow and frustrating."
review_lower = review.lower()

print("refund" in review_lower)
print("defective" in review_lower)
```

### Peeking Inside a String: Indexing and Slicing

Because a string is just a sequence of characters lined up in order, Python lets you reach in and grab individual characters, or entire chunks, using square brackets `[]`.

**Indexing** retrieves a single character by its position. Critically, Python starts counting positions at `0`, not `1` — so the *first* character of a string sits at position `0`, the second at position `1`, and so on. You can also count backward from the end using negative numbers: `-1` is always the *last* character, `-2` the second-to-last, and so on.

**Slicing** retrieves a whole chunk at once, using the pattern `string[start:end]`. This grabs every character starting at position `start`, up to — but **not including** — position `end`.

Let's use both to pull the exchange code and ticker out of a combined identifier string. Run the code cell below:

```python
ticker = "NASDAQ:AAPL"

print(ticker[0])
print(ticker[-1])
print(ticker[0:6])
print(ticker[7:])
```

A couple of details worth slowing down on:
* `ticker[0:6]` returned `NASDAQ` — six characters (positions 0 through 5), *not* including position 6 (the colon). This "stops one before the end number" rule trips up almost everyone the first time they see it, so it's worth remembering deliberately.
* `ticker[7:]` left the `end` position empty, which tells Python "go all the way to the end of the string." You can leave `start` empty the same way to mean "start from the very beginning."

You will use slicing constantly once you start working with structured text — stock tickers, transaction codes, dates written as plain text, and more all follow predictable patterns you can pull apart this way.

### Strings Are Immutable

Here's an important rule about strings that surprises a lot of newcomers: once a string is created, you **cannot change individual characters inside it**. In programming terms, strings are **immutable**.

This might sound like a strange limitation, but it is actually a very deliberate design choice that keeps your data safe and predictable — if a string gets passed around your program, you can always be certain it won't quietly change under you. Let's see what actually happens if you try to force it. Run the code cell below and read the error carefully:

```python
ticker = "AAPL"
ticker[0] = "Z"
```

We mentioned errors briefly back in the first lab, but now we get to see one in action for real. A good habit when reading a Python error is to read the traceback **from the bottom up**. The very last line is the actual error message: `TypeError: 'str' object does not support item assignment`. In plain English: we tried to replace the character (the "item") at position `0` with the letter `"Z"`, but this is not allowed for `str` objects. The `---->` arrow above it points to exactly which line of code triggered the problem — very handy once your code cells get longer than a line or two.

So how *do* you change a string, if you can't edit it in place? You already know the answer — you simply build a brand-new string and reassign it to the variable, exactly like we did earlier with `.upper()`, `.replace()`, and f-strings. None of those methods ever modify the original string; each one *returns a new string*, which we then store, often right back into the same variable name.

The built-in `str()` function works the same way. `str()` takes a value of pretty much any type and hands back a brand-new string version of it — the original value is left completely untouched. This comes in handy because Python's `+` operator refuses to concatenate a string with a number directly (try `"You own " + 150` and you'll get a `TypeError`, same as above). Wrapping the number in `str()` first fixes that. Run the code cell below:

```python
shares_owned = 150
price_per_share = 4.2

# print("You own " + shares_owned + " shares at $" + price_per_share + " each.")
# ^ this line would crash with a TypeError, because shares_owned and price_per_share are not strings

print("You own " + str(shares_owned) + " shares at $" + str(price_per_share) + " each.")
```

Notice that `shares_owned` is still the integer `150` after that line runs — `str()` didn't change it in any way. It simply handed back a brand-new string, `"150"`, which we then concatenated like any other piece of text. This is exactly why you'll see `str()` all over the place whenever you need to mix numbers into a message: `f"You own {shares_owned} shares"` (which we covered earlier) actually does this same `str()` conversion for you automatically behind the scenes!

We can even combine that same "build a new string" idea with slicing:

```python
ticker = "AAPL"
ticker = "Z" + ticker[1:]
print(ticker)
```

### Recap: Your String Toolkit

That was a lot of ground to cover! Before you move on to practice, here's a quick-reference table of everything we've learned. Bookmark this — you will come back to it constantly for the rest of the course.

| Tool | What It Does | Example |
|---|---|---|
| `f"...{variable}..."` | Builds a string with variables inserted directly | `f"Hello {name}"` |
| `+` | Concatenates (joins) strings together | `"Hello, " + name` |
| `\n`, `\t` | Escape sequences for a new line / a tab | `"Line 1\nLine 2"` |
| `"""..."""` | Triple quotes for a string that spans multiple lines | `"""Dear Sir,\n\n..."""` |
| `len(text)` | Counts the number of characters | `len("AAPL")` → `4` |
| `str(value)` | Converts a non-string value into a brand-new string | `str(150)` → `"150"` |
| `.upper()` | Converts every letter to UPPERCASE | `"aapl".upper()` → `"AAPL"` |
| `.lower()` | Converts every letter to lowercase | `"OK!".lower()` → `"ok!"` |
| `.title()` | Capitalizes the first letter of every word | `"hello world".title()` → `"Hello World"` |
| `.strip()` | Removes whitespace from the beginning and end | `"  hi  ".strip()` → `"hi"` |
| `.lstrip()` | Removes whitespace from the beginning only | `"  hi  ".lstrip()` → `"hi  "` |
| `.rstrip()` | Removes whitespace from the end only | `"  hi  ".rstrip()` → `"  hi"` |
| `.replace(old, new)` | Replaces every occurrence of `old` with `new` | `"cat".replace("c", "b")` → `"bat"` |
| `"word" in text` | Checks whether a substring appears (`True`/`False`) | `"cat" in "concatenate"` → `True` |
| `text[i]` | Indexing — grabs a single character at position `i` | `"AAPL"[0]` → `"A"` |
| `text[start:end]` | Slicing — grabs a chunk of characters | `"AAPL"[1:3]` → `"AP"` |

A key thing to remember: Every method in this table *returns a brand-new string* rather than changing the original (because as we mentioned earlier strings are immutable). That's why you'll almost always see them used together with an `=` to store the result, e.g. `cleaned_text = raw_text.strip().lower()`.
