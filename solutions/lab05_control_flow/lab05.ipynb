### Digital Skills and Business Decisioning Using Python

Every lab so far has been about *what* data looks like: strings, numbers, and Booleans (Labs 02 and 03), then lists, tuples, and dictionaries (Lab 04). But the code we've written still runs straight down the page, every line, every time.

**Control flow** is what changes that. It lets your program:

* make **decisions** (run certain lines only when some condition holds), and
* **repeat** work (run the same lines once for every item in a collection).

These two ideas, **conditionals** and **loops**, are the engine of almost every useful program. Once they're in place, we'll also introduce the **list comprehension**, a compact way to build a list that we held back on purpose until loops made sense.

### Part 1: Making Decisions with `if`

Back in Lab 03 you met the **Boolean**: a value that is either `True` or `False`, usually produced by a comparison like `price > 100`. On its own a Boolean just sits there. The `if` statement is what puts it to work: it runs a block of code **only when** its condition is `True`.

```python
if condition:
    # this indented block runs only if condition is True
```

Two things carry over from Lab 01: the line ends with a **colon** `:`, and everything that belongs "inside" the `if` is **indented** (four spaces). Python uses that indentation to know where the block starts and stops.

Run the code cell below:

````python
transaction_amount = 8500

if transaction_amount > 5000:
    print("Large transaction flagged for review.")

print("Done checking.")
````

The condition `8500 > 5000` is `True`, so the indented line ran. The last `print` is **not** indented, so it is not part of the `if`; it runs no matter what. Try changing `transaction_amount` to `200` and running again: the flag line is skipped, but `"Done checking."` still prints.

### `else`: the Other Path

Often you want one thing to happen when the condition is `True` and a *different* thing when it's `False`. That's what `else` is for.

Run the code cell below:

````python
transaction_amount = 3200

if transaction_amount > 5000:
    print("Large transaction flagged for review.")
else:
    print("Transaction approved automatically.")
````

Exactly one of the two blocks runs, never both. `else` has no condition of its own; it simply catches every case the `if` didn't.

### `elif`: More Than Two Outcomes

Real decisions often have several branches. `elif` (short for "else if") lets you test another condition when the previous ones were `False`.

Here we bucket a sentiment score (a number between -1 and 1, where negative means unhappy and positive means happy) into a label:

Run the code cell below:

````python
sentiment_score = 0.15

if sentiment_score > 0.2:
    print("Positive")
elif sentiment_score < -0.2:
    print("Negative")
else:
    print("Neutral")
````

Python checks the conditions **top to bottom**, runs the **first** one that is `True`, and skips everything else in the chain. Here `0.15 > 0.2` is `False` and `0.15 < -0.2` is `False`, so the `else` block runs and we get `Neutral`.

Because only the first match wins, **order matters**. You can have as many `elif` branches as you need, and the final `else` is optional (leave it out if there's nothing to do in the leftover case).

### Combining Conditions: `and`, `or`, `not`

In Lab 03 we promised these three keywords once `if` was on the table. They let you build a single condition out of several Booleans:

* `A and B` is `True` only when **both** `A` and `B` are `True`.
* `A or B` is `True` when **at least one** of them is `True`.
* `not A` flips a Boolean: `not True` is `False`.

Run the code cell below:

````python
price = 42.50
volume = 2_300_000

if price < 50 and volume > 1_000_000:
    print("Meets our buy criteria (cheap and actively traded).")

if not (price < 50):
    print("Too expensive.")
else:
    print("Price is within budget.")
````

The first `if` needed *both* halves to hold, and they did. In the second, `price < 50` is `True`, so `not (price < 50)` is `False`, and the `else` ran.

(The underscores in `2_300_000` are just a readability aid Python ignores; `2_300_000` and `2300000` are the same number.)

### Asking "Is It In There?": `in` inside an `if`

The `in` keyword you used on strings, lists, and dictionaries in Lab 04 produces a Boolean, so it slots straight into an `if`.

Run the code cell below:

````python
watchlist = ["AAPL", "MSFT", "GOOGL"]
ticker = "TSLA"

if ticker in watchlist:
    print(f"{ticker} is already tracked.")
else:
    print(f"{ticker} is not being tracked yet.")
````

### "Do I Have Any Data?": Truthiness

An `if` doesn't strictly need a comparison. Python treats some values as "as good as `True`" and others as "as good as `False`" when they land in an `if`:

* **Falsy** (act like `False`): `0`, `0.0`, an empty string `""`, an empty list `[]`, an empty dictionary `{}`, and `None`.
* **Truthy** (act like `True`): basically everything else, including any non-empty string, list, or dictionary.

This gives you a clean way to ask "is there anything here?":

Run the code cell below:

````python
reviews = []

if reviews:
    print(f"Analysing {len(reviews)} reviews.")
else:
    print("No reviews to analyse yet.")
````

`reviews` is an empty list, which is falsy, so the `else` ran. Add a string inside the brackets (`["good product"]`) and rerun: now the list is truthy and the first branch reports one review.

### Part 2: Repeating Work with Loops

You saw a preview of this in Lab 04: almost anything useful you do with a collection means touching **every item in turn**. Writing that out by hand (`print(tickers[0])`, `print(tickers[1])`, ...) is tedious and breaks the moment the list changes length. A **loop** does it for you.

### The `for` Loop

Run the code cell below:

````python
tickers = ["AAPL", "MSFT", "GOOGL"]

for ticker in tickers:
    print(f"Now analysing {ticker}")

print("Finished.")
````

Read it as English: *for each `ticker` in `tickers`, run the indented block*. On the first pass `ticker` holds `"AAPL"`, on the next `"MSFT"`, then `"GOOGL"`, and then the list is exhausted and the loop ends. The name `ticker` is your choice (`for t in tickers:` would work too), but a clear singular name makes the code readable.

Like `if`, the loop body is marked by the colon and the indentation. The final `print` is not indented, so it runs once, after the loop.

### The Accumulator Pattern

A very common job: walk through a list of numbers and add them all up. The trick is to create a "running total" variable **before** the loop and update it **inside**.

Run the code cell below:

````python
daily_returns = [0.4, -0.2, 1.1, -0.7, 0.9]

total = 0
for daily_return in daily_returns:
    total = total + daily_return

print(f"Total return over the period: {round(total, 1)}%")
````

`total` starts at `0`. Each pass adds the current value to it. After the loop, `total` holds the sum of the whole list.

`total = total + daily_return` is so common that Python has a shorthand: `total += daily_return` means exactly the same thing. You'll see `+=` everywhere from now on.

### An `if` Inside a `for`

The loop body is just ordinary code, so it can contain an `if`. Here we count how many headlines are longer than 25 characters:

Run the code cell below:

````python
headlines = [
    "Fed holds rates steady",
    "Tech earnings disappoint",
    "Oil prices climb on supply fears",
]

long_headlines = 0
for headline in headlines:
    if len(headline) > 25:
        long_headlines += 1

print(f"{long_headlines} of {len(headlines)} headlines are long.")
````

Two levels of indentation now: the `if` is inside the `for`, and the `long_headlines += 1` is inside the `if`. It only runs on passes where the condition holds. This "loop, test each item, count the ones that match" shape comes up constantly.

### Counted Loops: `range()`

Sometimes you don't have a list to walk; you just want to do something a fixed number of times. `range()` generates the numbers for you.

Run the code cell below:

````python
for i in range(5):
    print(f"Processing batch {i}")

print("---")

for year in range(2020, 2024):
    print(year)
````

`range(5)` gives `0, 1, 2, 3, 4`: it starts at `0` and **stops before** the number you pass (the same "counting from zero" idea as list indexes). `range(2020, 2024)` goes from the first number up to *but not including* the second. There's also a three-argument form, `range(0, 10, 2)`, which counts in steps of `2`.

When you're working with a list, prefer looping over it directly (`for ticker in tickers:`) rather than `for i in range(len(tickers)):`. It's shorter and harder to get wrong.

### The `while` Loop

A `for` loop runs a known number of times (once per item, or once per number in a range). A `while` loop runs an **unknown** number of times: it keeps going as long as its condition stays `True`, and stops as soon as the condition turns `False`.

Here we ask how many years of 5% annual growth it takes a balance to reach 1500:

Run the code cell below:

````python
balance = 1000
years = 0

while balance < 1500:
    balance = balance * 1.05
    years += 1

print(f"It took {years} years to pass 1500 (final balance: {round(balance, 2)}).")
````

The condition is checked **before** every pass. As long as `balance < 1500`, the body runs: grow the balance, add a year. Once `balance` finally reaches 1500 or more, the check fails and the loop stops.

The danger with `while` is the **infinite loop**: if nothing inside the body ever makes the condition `False`, it runs forever. Always make sure the body changes something the condition depends on. (If you do get stuck in one in Colab, press the stop button next to the cell.)

**Rule of thumb:** use `for` when you know what you're iterating over; use `while` when you're repeating until some condition changes.

### Cutting a Loop Short: `break` and `continue`

Two keywords give you finer control inside any loop:

* `break` exits the loop **immediately**, skipping any remaining items.
* `continue` skips the **rest of the current pass** and jumps straight to the next item.

Run the code cell below:

````python
prices = [101, 105, 98, 130, 102]

for price in prices:
    if price > 120:
        print(f"Outlier detected ({price}). Stopping scan.")
        break
    print(f"Checked {price}")
````

````python
returns = [0.4, -0.2, 1.1, -0.7, 0.9]

gain_days = 0
for daily_return in returns:
    if daily_return < 0:
        continue
    gain_days += 1

print(f"Number of positive days: {gain_days}")
````

In the first loop, `break` stopped the scan the moment a price above 120 appeared, so `102` was never checked. In the second, `continue` skipped the counting line on negative days, leaving `gain_days` counting only the positive ones.

### Looping Over a Dictionary

Dictionaries are built for iteration too. Looping over a dictionary directly gives you its **keys**:

Run the code cell below:

````python
word_counts = {"revenue": 12, "growth": 7, "loss": 3}

for word in word_counts:
    print(word)
````

Once you have the key, `word_counts[word]` gets the matching value, so you can print both:

````python
word_counts = {"revenue": 12, "growth": 7, "loss": 3}

for word in word_counts:
    print(f"{word}: {word_counts[word]}")
````

That works, but there's a cleaner way. `.items()` hands you **both** the key and the value on each pass, which you catch in two loop variables at once:

Run the code cell below:

````python
word_counts = {"revenue": 12, "growth": 7, "loss": 3}

for word, count in word_counts.items():
    print(f"'{word}' appeared {count} times")

total_words = 0
for count in word_counts.values():
    total_words += count
print(f"Total words counted: {total_words}")
````

`for word, count in word_counts.items():` is the standard way to walk a dictionary. `.values()` is there for when you only care about the values (here, to sum them with the accumulator pattern).

### Building a Dictionary With a Loop

Here's one of the most useful patterns in all of text analytics: given a list of words, count how many times each one appears, and store the result in a `{word: count}` dictionary.

Start with an empty dictionary. For each word, if you've seen it before, add one to its count; if not, start it at one.

Run the code cell below:

````python
words = ["risk", "growth", "risk", "return", "growth", "risk"]

counts = {}
for word in words:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print(counts)
````

The `if`/`else` handles the "first time I've seen this word" case separately from the "seen it before" case. This is common enough that `.get()` (from Lab 04) gives a one-line shortcut: `counts.get(word, 0)` returns the current count, or `0` if the word isn't in the dictionary yet.

Run the code cell below and confirm it gives the same answer:

````python
words = ["risk", "growth", "risk", "return", "growth", "risk"]

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
````

### Part 3: List Comprehensions

Now that loops make sense, we can introduce the shortcut we've been holding back.

One specific job comes up over and over: **make a new list by doing something to every item of an existing list**. Here's the long way, cleaning some column names into `snake_case`:

Run the code cell below:

````python
features = ["Word Count", "Avg Sentence Length", "Positive Words"]

snake_case = []
for feature in features:
    snake_case.append(feature.lower().replace(" ", "_"))

print(snake_case)
````

Four lines to build one list. A **list comprehension** does the same thing in one:

Run the code cell below:

````python
features = ["Word Count", "Avg Sentence Length", "Positive Words"]

snake_case = [feature.lower().replace(" ", "_") for feature in features]

print(snake_case)
````

Same result, one line. The shape is always:

```
[ EXPRESSION  for ITEM in COLLECTION ]
```

The `for ITEM in COLLECTION` part is exactly the loop you already know. The `EXPRESSION` at the front is what you want to *collect* for each item. Read it left to right: "give me `feature.lower().replace(...)` for each feature in features".

### Filtering: Adding an `if`

Put an `if` at the end and only the items that pass the test make it into the new list:

```
[ EXPRESSION  for ITEM in COLLECTION  if CONDITION ]
```

Run the code cell below:

````python
prices = [101, 105, 98, 130, 102, 145]
daily_returns = [0.4, -0.2, 1.1, -0.7, 0.9]

high_prices = [price for price in prices if price > 120]
gains_only = [r for r in daily_returns if r > 0]

print(high_prices)
print(gains_only)
````

`high_prices` kept only the values above 120; `gains_only` kept only the positive returns. Everything else was silently dropped.

### Comprehensions and Dictionaries

The collection you loop over can be `word_counts.items()`, so comprehensions work with dictionaries too. Here we pull out just the words that appear at least 5 times:

Run the code cell below:

````python
word_counts = {"revenue": 12, "growth": 7, "loss": 3, "risk": 1}

frequent_words = [word for word, count in word_counts.items() if count >= 5]

print(frequent_words)
````

There is also a **dictionary comprehension**, written with curly braces, for building a new dictionary: `{word: count * 2 for word, count in word_counts.items()}`. It's occasionally handy, but don't feel you need it; a plain `for` loop is always a fine choice.

### When *Not* to Use a Comprehension

Comprehensions are for one thing: building a list (or dict) with a simple transform and maybe a filter. Reach for a normal `for` loop when:

* the body needs several steps or its own `if`/`elif`/`else`,
* you're not building a collection at all (just printing, or updating a running total),
* the one-liner would be so long you can't read it at a glance.

Clear beats clever. A loop everyone can follow is better than a comprehension they have to decode.

### Recap: Your Control Flow Toolkit

| Tool | What It Does | Example |
|---|---|---|
| `if condition:` | Run the block only if the condition is `True` | `if score > 0.05:` |
| `elif condition:` | Test another condition when the ones above failed | `elif score < -0.05:` |
| `else:` | Run when no condition above was `True` | `else:` |
| `and` / `or` / `not` | Combine or flip Booleans | `price < 50 and volume > 1_000_000` |
| `in` / `not in` | Membership test (list, string, or dict keys) | `"AAPL" in watchlist` |
| `for item in collection:` | Repeat the block once per item | `for ticker in tickers:` |
| `range(n)` | The numbers `0` to `n - 1`, for counted loops | `for i in range(5):` |
| `while condition:` | Repeat while the condition stays `True` | `while balance < 1500:` |
| `break` / `continue` | Exit the loop / skip to the next item | `if price > 120: break` |
| `+=` | Add to a variable in place (accumulator) | `total += value` |
| `for k, v in d.items():` | Loop over a dictionary's key-value pairs | `for word, n in counts.items():` |
| `[expr for x in coll]` | Build a new list from a collection | `[p * 1.08 for p in prices]` |
| `[expr for x in coll if cond]` | Build a new list, keeping only some items | `[p for p in prices if p > 100]` |

The whole lab in one sentence: **`if` decides, `for` and `while` repeat, and a comprehension is a `for` loop that builds a list.**
