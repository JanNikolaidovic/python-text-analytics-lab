### **Strings and Text**

Almost every computer program exists to take in some kind of data, process it, and produce a useful result. Because data comes in so many different forms, programming languages classify it into specific **data types**.

The very first data type we will explore is the **string**. While strings might seem basic at first glance, they are incredibly versatile and you will use them constantly!

### What is a String?
At its core, a string is simply a sequence of characters (which can include letters, numbers, punctuation, or even spaces). In Python, absolutely anything wrapped inside quotation marks is instantly recognized as a string. 

Python is wonderfully flexible and allows you to use either double quotes (`" "`) or single quotes (`' '`) to define your strings. For example:
* `"Welcome to the data analytics lab."`
* `'Learning to code is a rewarding journey.'`

### The Power of Flexible Quotes
Why does Python give you two different ways to do the exact same thing? This clever flexibility allows you to easily include actual quotes or apostrophes *inside* your text without breaking your code! 

If your text contains a single quote (like an apostrophe), you simply wrap the entire string in double quotes. If you need to quote someone using double quotes, you wrap the string in single quotes. 

Here is how that works in practice:
* `'The manager said, "Please send me the report by Friday!"'`
* `"I'm really enjoying this chapter on data types."`
* `"The word 'algorithm' sounds intimidating, but it is just a set of instructions."`

### Changing Case in a String Using Methods

One of the most common (and easiest!) tasks when cleaning real-world business data is standardizing the capitalization of your text. 

Let's imagine you scraped some financial news headlines from a website, but the formatting is completely completely flat and lowercase. We can fix this using a Python **method**. A method is simply a specific action that Python can perform on a piece of data.

Run the code cell below to see how we can instantly format a messy news headline:

```python
news_headline = "central bank leaves interest rates unchanged"
print(news_headline.title())
```

If you ran the cell above, you should see the headline printed beautifully as:
`Central Bank Leaves Interest Rates Unchanged`

So, what exactly happened behind the scenes? Inside the `print()` command, we attached `.title()` to our variable. Let's break down that syntax:
* **The Dot (`.`):** The dot directly connects the variable to the method, telling Python to make the `title()` action act upon our `news_headline` string.
* **The Parentheses `()`:** Every method is followed by a set of parentheses. Often, methods need extra information to do their job, which you would place inside these parentheses. The `title()` method, however, knows exactly what to do (capitalize the first letter of every word) without any extra instructions, so its parentheses are left perfectly empty!

### The Upper and Lower Methods

Python offers several other handy methods for adjusting text case, which are absolute lifesavers in data analytics. Two of the most heavily used are `upper()` and `lower()`. 

* **The `upper()` method** is perfect for standardizing financial identifiers. For example, if a user types a stock ticker in lowercase, you can instantly convert it to the standardized uppercase format (like turning "aapl" into "AAPL").
* **The `lower()` method** is an absolute powerhouse for natural language processing. If you are analyzing thousands of customer reviews to gauge product sentiment, you can never trust the users' random capitalization. A common industry best practice is to convert all incoming review text to lowercase before analyzing it, ensuring that "OUTSTANDING!" and "outstanding!" are treated as the exact same word by your algorithm.

Let's see both of these in action! Run the code cell below:

```python
stock_ticker = "tsla"
print(stock_ticker.upper())

customer_review = "OUTSTANDING QUARTERLY EARNINGS!"
print(customer_review.lower())
```

### Combining or Concatenating Strings

When working with datasets, you will frequently need to combine multiple pieces of textual data. For example, imagine you pull a list of customers from a database where their first and last names are stored in completely separate columns, but you need to combine them to generate a personalized email. 

In Python, you can merge strings together using the plus symbol (`+`). This action is formally called **concatenation**.

Let's see how we can use the `+` symbol to combine a first name, a space, and a last name into a single variable. Run the code cell below:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

print(full_name)
```

If you ran the code, you should see the output `jane smith`. 

Concatenation is incredibly powerful when you want to automate complete messages, like a customized greeting for a marketing campaign. We can even combine concatenation with the string methods we learned earlier (like `.title()`) to ensure the final output is perfectly formatted!

Let's build a dynamic greeting. Run the following code block:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

print("Dear " + full_name.title() + ", welcome to our premium financial services!")
```

While the code above works perfectly and returns a beautifully formatted greeting (`Dear Jane Smith...`), writing long chains of text directly inside the `print()` function can get messy and hard to read. 

A much cleaner, professional approach is to build your entire message using concatenation, store that final message inside a new variable, and *then* print it. This makes your code much easier to manage.

Let's look at how we can store an automated email message into a single variable:

```python
first_name = "jane"
last_name = "smith"
full_name = first_name + " " + last_name

automated_email = "Dear " + full_name.title() + ", your monthly portfolio report is attached."
print(automated_email)
```
