### Part 1: Why Bother? A Problem Worth Solving

### Defining Your First Function

A function is created with the `def` keyword (short for "define"), a name you choose, parentheses `()`, and a colon. Everything indented underneath is the function's **body**: the code that runs each time the function is called.

```python
def function_name():
    # indented body: runs every time the function is called
```

Defining a function doesn't run it. Nothing happens until you **call** it, by writing its name followed by parentheses.

````python
def greet_analyst():
    print("Welcome to the trading floor.")
    print("Let's look at today's numbers.")

greet_analyst()
````

The `def` block only registers the function (Python reads and remembers it, nothing prints yet). Only the last line, `greet_analyst()`, actually calls it. Run the cell again after adding a second call, `greet_analyst()`, right below the first: the same two lines print a second time. Write the logic once, run it as many times as you like.

### Parameters: Giving a Function Input

`greet_analyst` always does exactly the same thing. Most of the time you want a function to work on *different* input each time you call it. A **parameter** is a placeholder name listed inside the parentheses when you define the function. An **argument** is the actual value you pass in when you call it.

```python
def function_name(parameter):
    # parameter behaves like a variable, but only inside this function
```

````python
def clean_headline(text):
    cleaned = text.strip().lower().rstrip(".")
    print(cleaned)

clean_headline("  Fed Holds Rates Steady.  ")
clean_headline("  TECH EARNINGS DISAPPOINT.  ")
clean_headline("  Oil Prices Climb On Supply Fears.  ")
````

One definition, three calls, three different results, and this is the fix for the copy-paste problem from the start of this part. `text` is a parameter: it holds no real value until the function is called, at which point Python fills it in with whatever argument you passed (first `"  Fed Holds...  "`, then the next string, then the next).

A function can take more than one parameter, separated by commas:

````python
def apply_discount(price, discount_percent):
    final_price = price - (price * discount_percent / 100)
    print(f"Final price: {round(final_price, 2)}")

apply_discount(100, 20)
apply_discount(49.99, 10)
````

Arguments are matched to parameters **by position**: the first value you pass (`100`) fills `price`, the second (`20`) fills `discount_percent`. Get the order backwards and the function still runs; it just computes the wrong thing, so pay attention to the order you defined the parameters in.

> **A function can also have a short docstring**: a one-line description right under `def`, wrapped in a pair of triple quotes, purely for human readers (it doesn't change how the function runs). You'll see this in other people's code, but it's entirely optional, and none of the exercises in this course require it.

### Avoiding Argument Errors

Every parameter without a default value is **required**: leave one out and Python refuses to run the function at all, rather than silently guessing. That's a good thing (it catches a mistake immediately instead of letting it cause confusion three cells later), but the error message can look intimidating the first time you see it.

````python
def apply_discount(price, discount_percent):
    return price - (price * discount_percent / 100)

# apply_discount(100)
# ^ this line would crash with:
# TypeError: apply_discount() missing 1 required positional argument: 'discount_percent' 
````

Read the last line of a traceback first: `TypeError` tells you *what kind* of mistake it is, and the rest of the message names the exact parameter you forgot. The same thing happens in reverse if you pass **too many** arguments (`"takes 2 positional arguments but 3 were given"`). Either way, the fix is the same: check the call against the `def` line and match the count.

### Part 2: Return Values, Getting Output Back

Every function you've written so far **prints** its result. That's fine for showing something to a human, but it's a dead end for the rest of your program: once a value is printed, Python can't use it again. If you want to **keep** a function's result (store it in a variable, feed it into another calculation, put it in a list), the function needs to `return` it.

````python
def add_tax(price):
    print(price * 1.08)

result = add_tax(100)
print(result)
````

`add_tax` printed `108.0`, and then the second `print(result)` printed `None`. `result` is `None` because `add_tax` never actually **returned** anything; printing inside a function is not the same as sending a value back out. Every function that doesn't explicitly `return` something hands back `None` by default.

````python
def add_tax(price):
    return price * 1.08

result = add_tax(100)
print(result)
````

Now `add_tax(100)` hands back the number `108.0`, which is stored in `result`. `return` is how a function communicates its answer to whatever code called it, exactly the way `len(my_list)` or `"hello".upper()` hand you back a value you can use.

`return` also ends the function immediately: any code after it, in that same call, never runs. This is useful for a function that should stop as soon as it knows its answer.

````python
def classify_sentiment(score):
    if score > 0.05:
        return "Positive"
    elif score < -0.05:
        return "Negative"
    return "Neutral"

print(classify_sentiment(0.4))
print(classify_sentiment(-0.6))
print(classify_sentiment(0.0))
````

As soon as one branch hits its `return`, the function stops there and hands that string back; the branches below it never even run. Notice there's no `else` before the last `return`: if neither the `if` nor the `elif` fired, Python simply falls through to that final line.

### A Note on Scope: What Happens Inside, Stays Inside

A variable created **inside** a function only exists inside that function; this is called **local scope**. The moment the function finishes running, that variable is gone. A variable created outside any function (everything you've used so far in these labs) lives in the **global scope** instead.

````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(10, 3)
print(result)

# print(total)
# ^ this line would crash with a NameError: total only exists inside calculate_total
````

`total` is local: created fresh every time `calculate_total` runs, and destroyed the moment it returns. Only `result`, the value that came back out through `return`, survives outside. This is a feature, not a limitation: every function's internal variables are private, so you can reuse a name like `total` in ten different functions without them interfering with each other.

A function *can* read a global variable without it being passed in as a parameter:

````python
tax_rate = 0.08

def add_tax(price):
    return price * (1 + tax_rate)

print(add_tax(100))
````

That works, but it's fragile: `add_tax`'s behavior now secretly depends on whatever `tax_rate` happens to be, elsewhere in the notebook. And it only goes one way: assigning to a variable *inside* a function never changes a variable of the same name outside it.

````python
counter = 0

def try_to_change_counter():
    counter = 5  # this creates a NEW local variable; it does not touch the outer one
    print("Inside the function:", counter)

try_to_change_counter()
print("Outside the function:", counter)
````

Even though both are named `counter`, they're two different boxes. The assignment inside the function creates a brand-new local variable that exists only during that call; the global `counter` outside is never touched, which is why it still prints `0`. If you want a function's result available outside, don't rely on reassigning a global: `return` it, and store the result in a variable, the pattern you've been using all lab.

### But Lists (and Dictionaries) Work Differently

The `counter` example worked the way it did because numbers (and strings) can't be changed in place: `counter = 5` inside the function threw away the old value and pointed the local name at a brand-new one. Lists and dictionaries are **mutable**: they can be changed in place, and a parameter that receives one is pointing at the *exact same* list the caller has, not a copy.

````python
def add_bonus_points(scores):
    scores.append(100)

leaderboard = [85, 92, 78]
add_bonus_points(leaderboard)

print(leaderboard)
````

There's no `return` anywhere, yet `leaderboard` ends up with four items. `add_bonus_points` never reassigned its parameter (that would behave like the `counter` example, and leave `leaderboard` untouched); it called `.append()` on it, which changes the list itself, in place. Since `scores` inside the function and `leaderboard` outside are two names for the *same* list, the change is visible everywhere.

> **Reassigning a parameter (`scores = [...]`) never affects the caller; mutating it (`.append()`, `.pop()`, `scores[0] = ...`, or setting a dictionary key) always does.** Keep this distinction in mind once you start writing functions that work on lists and dictionaries, since it's easy to assume `return` is the only way for a function to hand back a change.

### Part 3: Flexible Input and Multiple Answers

Sometimes a parameter has an obvious "usual" value that you only want to override occasionally. A **default value** lets you skip the argument entirely when the usual value is fine.

````python
def apply_discount(price, discount_percent=10):
    return price - (price * discount_percent / 100)

print(apply_discount(100))
print(apply_discount(100, 25))
````

Called with just `100`, `discount_percent` falls back to its default, `10`. Passing a second argument, `25`, overrides the default for that call. A parameter with a default is always optional; parameters without one (like `price` here) are still required on every call.

One rule to remember: every parameter with a default must come *after* every parameter without one. `def apply_discount(discount_percent=10, price):` is not allowed.

So far every argument has been matched to a parameter **by position** (first argument, first parameter, and so on). You can also pass arguments **by name**, called a keyword argument, which frees you from remembering the exact order.

````python
def describe_stock(ticker, price, volume):
    print(f"{ticker}: ${price}, volume {volume}")

describe_stock("AAPL", 189.5, 45_000_000)
describe_stock(ticker="AAPL", volume=45_000_000, price=189.5)
````

Both calls print the same thing. The second spells out `ticker=`, `price=`, `volume=` explicitly, so order no longer matters: Python matches by name instead of position. Keyword arguments also let you override just one default and leave the rest alone:

````python
print(apply_discount(100, discount_percent=25))
````

### Returning More Than One Value

A function isn't limited to a single answer. Separate multiple values with a comma after `return`, and Python bundles them into a tuple (the data structure from Lab 04). You can unpack that tuple into several variables at once, right at the call site.

````python
def price_range(prices):
    return min(prices), max(prices)

daily_prices = [101.2, 100.8, 103.5, 98.7, 104.0]

lowest, highest = price_range(daily_prices)
print(f"Range: {lowest} to {highest}")
````

`price_range` returns two values at once; `lowest, highest = price_range(daily_prices)` unpacks them into two variables in a single line, the same unpacking trick you used on tuples in Lab 04. If you only capture it in one variable instead (`result = price_range(daily_prices)`), you'd get the whole tuple back: `(98.7, 104.0)`.

### Returning a Dictionary

A tuple works fine for two or three values, but once there are several, remembering "the second item is volume, the third is the sector" gets error-prone. When each value deserves a clear label, return a **dictionary** instead (the data structure from Lab 04): `return` isn't limited to numbers and tuples, any data type can follow it.

````python
def build_stock_record(ticker, price, volume):
    return {"ticker": ticker, "price": price, "volume": volume}

record = build_stock_record("AAPL", 189.5, 45_000_000)
print(record)
print(record["price"])
````

`build_stock_record` packages three related values into a single dictionary and returns it in one step. Reading `record["price"]` afterwards is a lot clearer than remembering that "price is the second item in the tuple", especially as the number of fields grows.

### Part 4: Functions and Loops, the Real Payoff

A function that takes one item and returns one result is exactly what a `for` loop needs to process a whole collection. This combination, a small function plus a loop that calls it on every item, is one of the most common patterns in text analytics.

````python
def clean_headline(text):
    return text.strip().lower().rstrip(".")

raw_headlines = [
    "  Fed Holds Rates Steady.  ",
    "  TECH EARNINGS DISAPPOINT.  ",
    "  Oil Prices Climb On Supply Fears.  ",
]

cleaned_headlines = []
for headline in raw_headlines:
    cleaned_headlines.append(clean_headline(headline))

print(cleaned_headlines)
````

This is the Part 1 problem, solved properly: `clean_headline` holds the recipe exactly once, and the loop calls it once per headline, collecting every result into a new list with the accumulator pattern from Lab 05.

Since this is "build a new list by doing something to every item", it's also a natural fit for the list comprehension you met in Lab 05:

````python
cleaned_headlines = [clean_headline(headline) for headline in raw_headlines]
print(cleaned_headlines)
````

Same result, one line: the comprehension calls `clean_headline` on every item and collects the returned values into a new list. Whichever form you reach for, the key idea is the same: write the logic once as a function, then let a loop (or comprehension) apply it everywhere it's needed.

A function can also be the **condition** a loop checks, when it returns a Boolean:

````python
def mentions_earnings(headline):
    return "earnings" in headline.lower()

headlines = [
    "Apple earnings beat expectations",
    "Oil prices rise",
    "Tesla Earnings call scheduled",
    "Fed holds rates",
]

for headline in headlines:
    if mentions_earnings(headline):
        print(f"Match: {headline}")
````

`mentions_earnings` returns `True` or `False`, so it slots directly into the loop's `if`, exactly like any other Boolean condition from Lab 05.

### Functions With a while Loop, and One Job Per Function

Functions work with `while` loops exactly as well as they do with `for` loops. This example also shows a habit worth building early: instead of one function trying to do everything, split the work into two smaller functions, each with **one job**. Here, one function works through a queue of pending headlines, and a second one reports what got done.

````python
def process_headlines(pending, processed):
    while pending:
        headline = pending.pop()
        processed.append(headline.upper())

def show_processed(processed):
    print("Processed headlines:")
    for headline in processed:
        print(f"- {headline}")

pending_headlines = ["fed raises rates", "tech stocks rally", "oil prices dip"]
processed_headlines = []

process_headlines(pending_headlines, processed_headlines)
show_processed(processed_headlines)
````

`process_headlines` uses a `while` loop the same way Lab 05 did: keep going as long as `pending` (a list, which is truthy while it has items) isn't empty, `.pop()` one item at a time, and push the result into `processed`. Notice neither function `return`s anything: `pending` and `processed` are both lists, so the mutations inside the function (`.pop()`, `.append()`) are visible on the exact same lists back in the calling code, the mutability rule from Part 2. Splitting the job in two, one function that does the processing and one that reports the result, also makes each function easier to read, test, and reuse on its own.

### Part 5: Functions With a Flexible Number of Arguments

Every function so far has had a fixed number of parameters. Sometimes you genuinely don't know in advance how many values a caller will want to pass in, for example, a function that totals up however many numbers you hand it, two on one call and five on the next. Python has two special parameter forms for exactly this, and you'll run into both when reading other people's code.

#### `*args`: Any Number of Positional Values

A parameter written with a single `*` in front collects every extra positional argument into a tuple.

````python
def total_of(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(total_of(10, 20))
print(total_of(4.5, 1.2, 3.3, 0.8))
````

`*numbers` packs whatever you pass, two values the first time, four the next, into a single tuple called `numbers` inside the function. From there it's an ordinary loop, the accumulator pattern from Lab 05. You'll recognize this `*` when you meet it elsewhere: built-in functions like `print()` accept any number of values using the same trick.

#### `**kwargs`: Any Number of Keyword Values

Two asterisks do the same thing for **keyword** arguments, collecting them into a dictionary instead of a tuple.

````python
def describe_company(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

describe_company(name="Acme Corp", sector="Industrials", employees=1200)
````

`**details` gathers `name="Acme Corp"`, `sector="Industrials"`, and `employees=1200` into one dictionary, then the function loops over it with `.items()`, the same pattern from Lab 05. You'll see `**kwargs` constantly in library code (pandas and plotting functions lean on it heavily) even if you rarely need to write your own; recognizing what it means is the main goal here.

### Part 6: Reusing Other People's Functions With import

Every function you've written this lab has lived inside this one notebook. In a bigger project you'd often save related functions in their own `.py` file, called a **module**, and pull them into other files with an `import` statement. You're unlikely to split a Colab notebook into multiple files, but you'll use `import` constantly anyway, since it's exactly how you get access to functions other people have already written: Python's standard library today, and packages like pandas very soon.

````python
import math

print(math.sqrt(64))
print(math.ceil(4.1))
````

`import math` makes every function inside the `math` module available through **dot notation**: `math.sqrt(...)`, `math.ceil(...)`. This is exactly the pattern you'll use later for `import pandas as pd`, followed by `pd.read_csv(...)`.

If you only need one or two functions from a module, `from module import function` brings just those names in directly, so you can call them without the module prefix.

````python
from statistics import mean

daily_returns = [18.0, 22.5, 19.0, 26.5]
print(mean(daily_returns))
````

Compare this to the `average()` function you wrote by hand in Part 2's practice: Python's standard library (and, later, pandas) already provides carefully tested versions of most calculations you'd want, `statistics.mean` here. Writing your own version first is a great way to *understand* what the calculation actually does; once you do, reach for the library version in real work.

Finally, `import module as alias` gives a module a short nickname, handy when the full name is long or you'll type it often. This is exactly why you'll almost always see `import pandas as pd` rather than spelling out `pandas` every time.

````python
import statistics as stats

print(stats.mean(daily_returns))
print(stats.median(daily_returns))
````

### Recap: Your Functions Toolkit

| Tool | What It Does | Example |
|---|---|---|
| `def name(): ...` | Defines a function; nothing runs until it's called | `def greet(): ...` |
| `name()` | Calls a function | `greet()` |
| `def name(param): ...` | Defines a parameter the caller fills in | `def clean(text): ...` |
| missing/extra arguments | Raises a `TypeError` naming the parameter that's missing or unexpected | `missing 1 required positional argument` |
| `return value` | Sends a value back out and ends the function | `return price * 1.08` |
| `None` | What a function returns if you never write `return` | `result = greet()` (`None`) |
| `def name(param=default): ...` | Makes a parameter optional | `def tax(price, rate=0.08):` |
| `name(param=value)` | A keyword argument, matched by name, not position | `describe(ticker="AAPL")` |
| local scope | Variables created inside a function disappear once it returns | `total` inside `calculate_total` |
| global scope | Variables created outside any function, readable (not writable) from inside | `tax_rate` used inside `add_tax` |
| mutable arguments | Lists and dictionaries passed in can be changed in place, no `return` needed | `scores.append(100)` |
| `return a, b` | Returns multiple values as a tuple | `return min(x), max(x)` |
| `a, b = func(...)` | Unpacks a multi-value return into separate variables | `low, high = price_range(prices)` |
| `return {...}` | Returns a dictionary of labeled values | `return {"ticker": t, "price": p}` |
| function + loop | Apply one function's logic to every item in a collection | `[clean(t) for t in texts]` |
| `def name(*args):` | Collects any number of positional arguments into a tuple | `def total_of(*numbers):` |
| `def name(**kwargs):` | Collects any number of keyword arguments into a dictionary | `def describe(**details):` |
| `import module` | Makes a module's functions available via `module.function()` | `import math` |
| `from module import x` | Imports just what you need, no module prefix | `from statistics import mean` |
| `import module as alias` | Gives a module a short nickname | `import pandas as pd` |

The whole lab in one sentence: **a function packages logic under a name, `return` is how it hands back an answer (a single value, several at once, or a dictionary), and once you can write your own, `import` is how you start reusing everyone else's.**
