## **2. The Basics**

### **What really happens when you run a code cell?**

Let's take a closer look at what Python actually does when you run a cell containing code. 

Think of writing code as giving a direct set of instructions to the computer. Every time you type a line of code, you are commanding Python to perform a specific task. Python takes that instruction, processes it behind the scenes, and (given that there are no mistakes in your code) returns the result you asked for.

In fact, Python does quite a bit of heavy lifting even when executing a single, simple line of code:



````python
#Run this code block.
print("Hello course taker!")
````

After running the code cell above, you should see the following output appear directly below it: `"Hello course taker!"`

When you execute a code cell, the notebook sends your command to the **Python interpreter**. Think of the interpreter as a translator: it reads through your program, evaluates the syntax, and determines exactly what each word means. For example, when the interpreter sees the word `print` followed by parentheses, it knows its specific job is to display whatever is inside those parentheses onto your screen.

You might have also noticed that the text inside the code cell is not just plain black. As you write, the notebook automatically highlights different parts of your program in different ways. For instance, it recognizes that `print` is a built-in command (known as a **function**) and gives it a specific color. It also recognizes that `"Hello course taker!"` is a textual data value (known as a **string**) rather than a command, and displays it in an entirely different color. 

This incredibly helpful feature is called **syntax highlighting**. As you start writing longer and more complex code, these visual color clues will become your best friend for understanding your code's structure and spotting typos quickly!

### **Variables**
Let's try using a variable, by taking one of the previous examples. Run the 
code cell bellow. You should see the same output as you saw previously.

````python
message = "Hello course taker!" 
print(message)
````

So, what exactly did we just do? We created a new variable named `message`. 

In Python, you can think of a **variable** as a labeled storage container connected to a specific **value** (which is simply the data or information you want to save and use later). In this specific case, the value we stored inside our container is `"Hello course taker!"`. Because this value is made up of text, programmers refer to it as a **string**.

Adding a variable makes a little more work for the python interpreter. 
When it processes the first line, it associates the variable message with the
"Hello course taker!" text. When it reaches the second line, it prints the value
associated with message to the screen. Let's do it again, but more times!

````python
message = "Hello course taker!" 
print(message)

message = "You can do this!" 
print(message)
````

As you saw in the previous cell's output, we used the exact same variable `message` twice, but printed two completely different results! 

Why did this happen? Because you can change the value of a variable in your code at any time. If you think back to our labeled container analogy, assigning a new value simply removes the old data from the box and replaces it with the new information. Because Python reads your code from top to bottom, it will always keep track of (and use) a variable's most current value.

### **Naming and Using Variables**

When creating variables in Python, there are a few strict rules and helpful guidelines you need to follow. Breaking the strict rules will cause Python to throw an error, while ignoring the guidelines will simply make your code harder for you (and others) to read. 

Keep these golden rules in mind when naming your variables:

* **Letters, Numbers, and Underscores Only:** Variable names can only contain these three types of characters. They can start with a letter or an underscore, but they **cannot** start with a number. For instance, you can name a variable `message_1`, but `1_message` will break your code.
* **No Spaces Allowed:** You cannot use spaces in variable names. Instead, use underscores to connect your words. For example, `hello_message` works perfectly, but `hello message` will cause a syntax error.
* **Avoid Python Keywords:** Do not use built-in Python function names or keywords as your own variable names. For example, never name a variable `print`, because Python has already reserved that word for a very specific job!
* **Keep it Short but Descriptive:** A good variable name tells you exactly what data it holds. For example, `name` is much better than just `n`, `student_name` is better than `s_n`, and `name_length` is much easier to read than `length_of_a_persons_name`.
* **Watch Out for Confusing Letters:** Be careful when using the lowercase letter `l` (el) and the uppercase letter `O` (oh). Depending on the font you use, these can easily be confused with the numbers `1` and `0`!

> 💡 **Best Practice Note:** At this stage in your journey, all your variable names should be entirely lowercase. While using uppercase letters won't cause an error, capital letters are generally reserved by developers for special types of variables that we will explore later in the course.

### **Avoiding Name Errors (And Embracing Mistakes!)**

As you begin writing your own code, you are going to make mistakes. This is completely natural! In fact, every single programmer makes mistakes every single day, no matter the seniority. Overall, it is not about writing perfect code on the first try, but rather knowing how to efficiently read, understand, and fix those inevitable errors. 

Let's look at a very common error you are likely to encounter early on and learn how to resolve it. 
Run the code cell bellow.

````python
message = "Let's generate an error!"
print(mesage)
````

When an error (often called a "bug") occurs in your program, the Python interpreter doesn't just crash in silence; it actually does its absolute best to help you figure out what went wrong. When a program cannot run successfully, the interpreter provides a helpful report called a **traceback**.

> 🔍 **What is a traceback?** > Think of it as a breadcrumb trail. It is a detailed record of exactly where the interpreter got confused and ran into trouble while trying to execute your code.



So, what is that traceback actually telling you? Let's break it down:

* **Where the error happened:** The output specifically points to **line 2** of the executed code cell. Python does this to help you track down the exact location of the bug instantly.
* **What kind of error it is:** It explicitly reports a `NameError` and explains that the name `mesage` is not defined. 

A `NameError` usually means one of two things: we either forgot to assign a value to the variable before trying to use it, or (as in this case) we simply made a spelling mistake. Python is looking for a container named `mesage`, but it cannot find one!

> 💡 **Fun Fact:** Modern versions of Python are remarkably smart! If you make a typo, Python will scan your code for similar variable names. If it finds a match, it will actually ask, *"Did you mean: 'message'?"* right inside the error output to help you fix it faster. If it can't guess what you meant, it will just throw the standard error.


In the previous example, we intentionally omitted the letter 's' from the variable name `message` on the second line. 

It is important to understand that the Python interpreter doesn't know English (it will never spellcheck your vocabulary). However, it *does* care deeply about **consistency**. It expects you to spell a variable name exactly the same way every single time you use it in your code. 

For instance, let's see what happens if we make a typo in the very first line where the variable is actually defined. Run the code cell below to find out!

````python
mesage = "That is the last hello for today." 
print(mesage)
````

In this case the program runs successfully. Because the variable names match perfectly, Python sees absolutely no issue. 

Remember: programming languages are incredibly strict about consistency, but they do not care at all about proper English spelling or grammar. As long as you spell your custom variable names exactly the same way every time, you are good to go.

### **Comments**


One of the most important habits you can build as a programmer is leaving **comments** in your code. 

Imagine you write a very clever piece of code to solve a specific problem today. A few weeks pass, you look back at it, and suddenly you have no idea what it does or why you wrote it! To prevent this, you can use the hash symbol (`#`) inside a code cell to leave helpful notes for yourself. Think of it like scribbling notes in the margins of a textbook.

Comments are also crucial because you are rarely the only person reading your code. Colleagues, managers, and instructors will often need to review it too. Good comments explain the *what* and the *why* of your logic so anyone can easily follow your thought process.
 
**⚠️ Be careful:** The Python interpreter completely ignores everything to the right of the `#` symbol on a given line. It treats that text purely as a note and will not run it. This means if you put a `#` in front of an actual Python command, that code is effectively disabled (or "commented out") and will not execute! 
 
Run the code cell below to see this in action—notice how it produces absolutely no output.

## **3. Python Language Basics**

So far you've been running code and creating variables — but you haven't yet seen what actually makes Python *look* like Python. Python's design leans hard into readability and clarity; some programmers half-jokingly call it "executable pseudocode," because well-written Python often reads almost like plain English instructions. Let's look at two of the habits that give the language this feel.

### **Indentation, Not Braces**

Many programming languages — Java, C++, and R among them — use curly braces `{ }` to mark where a block of code starts and ends. Python throws the braces out entirely and uses **whitespace** instead.

A colon (`:`) signals the start of a new indented block. Every line that belongs to that block has to be indented by *exactly* the same amount, and the block ends the moment the indentation drops back to where it started. Here's a classic example. You don't need to understand every keyword yet — `for` and `if` are coming in a future lab — just notice how the indentation visually groups the lines that belong together:

```python
for number in numbers:
    if number < pivot:
        smaller.append(number)
    else:
        larger.append(number)
```

Each `:` is immediately followed by a new, more-indented block, and the code drops back a level once that block is finished. Love it or hate it, indentation *is* the syntax in Python — there's no bracket to fall back on, and no way to opt out of it.

````python
temperature = 30

if temperature > 25:
    print("It's a hot day!")
else:
    print("It's a mild day.")
````

Run the cell above. Even though you haven't formally learned `if` statements yet, you can probably already guess what happened: because `temperature` is greater than `25`, Python ran the indented line right under `if` and skipped the one under `else`. We'll unpack `if` statements properly in a future lab — for now, just focus on the pattern: **colon, then indent**.

> 💡 **Tip:** Stick to **four spaces** per indentation level — it's the convention almost every Python programmer follows. You don't even have to count them yourself: Colab (and Jupyter notebooks in general) automatically indents four spaces for you every time you start a new line after a colon, and quietly turns any Tab key press into spaces too.

### **Watch Your Indentation: A New Kind of Error**

Back when we practiced spotting a `NameError`, we saw that Python doesn't fail silently — it tells you what went wrong. Inconsistent indentation is another extremely common beginner mistake, and Python has a dedicated error just for it. Run the cell below to meet it.

````python
temperature = 30
if temperature > 25:
    print("It's a hot day!")
     print("Grab some water!")
````

This time the interpreter raises an `IndentationError` instead of a `NameError`. Just like before, the traceback points you straight to the line that confused it — Python expected that second `print()` to line up exactly with the one above it, and it didn't.

The fix is simple: make sure every line inside the same block starts at exactly the same column. Fix it below and re-run to confirm you get both lines of output.

````python
temperature = 30
if temperature > 25:
    print("It's a hot day!")
    print("Grab some water!")
````

### **Semicolons? Rarely.**

Unlike languages that require a semicolon (`;`) at the end of every line, Python statements don't need one at all. You *can* use semicolons to squeeze multiple statements onto a single line:

```python
a = 5; b = 6; c = 7
```

...but doing this is generally considered poor style in Python, since it makes your code harder to scan at a glance. Stick to one statement per line — your future self (and anyone reading your code later) will thank you.

### **Everything in Python Is an Object**

Here's a fact that will quietly explain a lot of Python's behavior as this course goes on: **everything in Python is an object**. Every number, every string, every function, and every more complex structure you'll meet later on all live inside the interpreter as an "object" — its own little package that carries both a **type** (what kind of thing it is) and the actual data itself.

You've already been creating objects without necessarily thinking of them that way! Every time you assigned a variable, you attached a label to one of these objects. Python's built-in `type()` function lets you ask any object what kind of object it is. Try it below:

````python
value = "Hello course taker!"
print(type(value))

count = 5
print(type(count))
````

`str` is short for **string** — Python's type for text. `int` is short for **integer** — Python's type for whole numbers. You'll meet several more built-in types, and learn exactly how to work with each one, in the next lab.

For now, the big takeaway is this: because *everything* is an object with this same consistent structure, Python ends up treating data, functions, and more in a similarly flexible way. That consistency is a big part of what makes Python so pleasant to work with.
