### What Is a List?

A **list** is an ordered collection of values, gathered together in a single variable. You create one by wrapping the values in **square brackets** `[ ]` and separating them with commas:

```python
features = ["word_count", "avg_sentence_length", "positive_words", "negative_words"]
```

A few things to notice:

* The list is **ordered** (the items stay in the order you wrote them, and each has a fixed position).
* A list can hold **any** data type (strings, numbers, Booleans, even other lists).
* In practice, the items in a list are usually the *same kind of thing*: a list of tickers, a list of prices, a list of words. Nothing forces this, but it's what makes a list useful.

Run the code cell below to create two lists and print them:

````python
features = ["word_count", "avg_sentence_length", "positive_words", "negative_words"]
daily_prices = [101.2, 100.8, 103.5, 102.1, 104.0]

print(features)
print(daily_prices)
````

Python prints the list exactly as you'd expect (square brackets, items in order, comma-separated). The whole collection lives inside one variable, but the individual values are still in there, and we can reach in and grab any one of them.

### Accessing Elements in a List

To pull a single value out of a list, write the list's name followed by the item's **position number** in square brackets. That position number is called the **index**.

Run the code cell below:

````python
features = ["word_count", "avg_sentence_length", "positive_words", "negative_words"]

print(features[0])
print(features[1])
print(features[3])
````

### Why Do We Start Counting at 0?

Look closely at that last cell. `features[0]` gave us `"word_count"` (the **first** item). `features[1]` gave us the **second**. This catches almost everyone out at first, so it's worth slowing down.

In Python, the index is not a "1st, 2nd, 3rd" count. It's the **distance from the start of the list**. The first item is `0` steps from the beginning, the second item is `1` step along, and so on. So the index of the *n*-th item is always *n* minus 1.

This is not a Python quirk. It's the norm across most of programming:

| Starts counting at `0` | Starts counting at `1` |
|---|---|
| Python, JavaScript, Java, C, C++, C# | R, MATLAB, Julia, Fortran |

If you've used **R** for statistics, this is a real adjustment. In R, `features[1]` gives you the *first* element:

```r
# R
features <- c("word_count", "avg_sentence_length", "positive_words")
features[1]   # -> "word_count"
```

In Python, that same `features[1]` gives you the **second** element, and `features[0]` is the first. When you move between the two languages, the off-by-one difference is one of the most common sources of quiet bugs, so keep it front of mind.

Run the code cell below to see the idea spelled out:

````python
features = ["word_count", "avg_sentence_length", "positive_words", "negative_words"]

# The index is "how many steps from the start", not a 1, 2, 3 count.
first_feature  = features[0]   # 0 steps from the start  -> the 1st item
second_feature = features[1]   # 1 step  from the start  -> the 2nd item

print(first_feature)
print(second_feature)
````

### Reaching the End: Negative Indexes

Often you want the *last* item in a list, but you don't know (or don't want to count) how long the list is. Python lets you index from the **end** using negative numbers:

* `list[-1]` is the last item
* `list[-2]` is the second-to-last
* ...and so on

Run the code cell below:

````python
features = ["word_count", "avg_sentence_length", "positive_words", "negative_words"]

print(features[-1])
print(features[-2])
````

`features[-1]` is the cleanest way to say "the last item, whatever it is". You'll reach for it constantly, and it saves you from having to work out the length first.

### Using Individual Values from a List

An item pulled out of a list is just an ordinary value (a string or a number like any other). You can drop it into an f-string, do arithmetic with it, or call methods on it, exactly as you learned in the previous labs.

Run the code cell below:

````python
tickers = ["AAPL", "MSFT", "GOOGL"]
print(f"The first company on our watchlist is {tickers[0]}.")

daily_prices = [101.2, 100.8, 103.5]
price_change = round(daily_prices[2] - daily_prices[0], 2)
print(f"Price moved by {price_change} over the period.")
````

Nothing new here except *where* the values came from. `tickers[0]` is a string, so it slots into the f-string. `daily_prices[2]` and `daily_prices[0]` are floats, so we can subtract them. We wrap the result in `round()` (from the last lab) to trim the trailing decimals that a subtraction of floats can leave behind. Once a value is out of the list, the list stops mattering; you are just working with a number or a string.

### Changing an Element

A list is **mutable** (a fancy word meaning you can change its contents after you've created it). To replace an item, assign a new value to its index, the same way you assign to a variable:

Run the code cell below:

````python
watchlist = ["AAPL", "MSFT", "GOOGL"]
watchlist[1] = "NVDA"
print(watchlist)
````

`watchlist[1] = "NVDA"` reached into position `1` and swapped `"MSFT"` for `"NVDA"`. The rest of the list is untouched. (Strings, by contrast, can't be changed like this: `headline[0] = "X"` is an error. This is a real difference between the two types.)

### Adding Elements: `append()` and `insert()`

Two methods add new items to an existing list:

* **`.append(x)`** adds `x` to the **end** of the list. This is by far the most common way to grow a list.
* **`.insert(i, x)`** slides `x` into position `i`, shifting everything from that point onward one place to the right.

Run the code cell below:

````python
watchlist = ["AAPL", "MSFT", "GOOGL"]

watchlist.append("NVDA")
print(watchlist)

watchlist.insert(0, "AMZN")
print(watchlist)
````

A very common pattern is to start with an **empty list** and append to it as you go, building results up one piece at a time. An empty list is just `[]`.

Run the code cell below:

````python
tickers = []
tickers.append("AAPL")
tickers.append("MSFT")
print(tickers)
````

### Removing Elements: `del`, `pop()`, and `remove()`

There are three ways to take something *out* of a list, and the right one depends on what you know about the item you want gone:

| Tool | You specify... | Gives the value back? | Typical use |
|---|---|---|---|
| `del list[i]` | the **position** | no | You know where it is and you're done with it |
| `list.pop(i)` | the **position** (default: the last) | **yes** | You want to remove it *and* keep using it |
| `list.remove(x)` | the **value** | no | You know *what* to remove, not where it is |

**`del`: remove by position, and it's gone.**

````python
watchlist = ["AMZN", "AAPL", "NVDA", "MSFT", "GOOGL"]
del watchlist[0]
print(watchlist)
````

**`pop()`: remove by position, but hand the value back.** With no index it removes the *last* item; with an index it removes that one. This is the tool to use when the removed item still matters (for example, pulling the next task off a to-do list to work on it).

````python
watchlist = ["AAPL", "NVDA", "MSFT", "GOOGL"]

last_removed = watchlist.pop()
print(last_removed)
print(watchlist)

second_removed = watchlist.pop(1)
print(second_removed)
print(watchlist)
````

Notice `pop()` did two jobs at once: it changed the list *and* returned the value it took out, so we could store it in `last_removed` and print it. `del` can't do that; after `del`, the value is simply gone.

**`remove()`: remove by value.** When you know *what* you want to delete but not its position, `remove()` searches the list and deletes the **first** matching item.

````python
watchlist = ["AAPL", "NVDA", "MSFT", "GOOGL"]
watchlist.remove("NVDA")
print(watchlist)
````

Two things to remember about `remove()`:

* It only removes the **first** occurrence. If `"NVDA"` appeared twice, the second one would stay.
* If the value isn't in the list at all, Python raises a `ValueError`. If you're not sure an item is present, you'll want to check first, a job for the `in` keyword you met with strings (`"NVDA" in watchlist`), which works on lists too.

### Organizing a List: `sort()` and `sorted()`

You'll frequently want a list in order (prices low to high, names alphabetically, dates oldest to newest). Python gives you two ways to do it, and the difference between them matters.

**`.sort()` sorts the list *in place*, permanently.** The original order is lost.

Run the code cell below:

````python
tickers = ["MSFT", "AAPL", "GOOGL", "AMZN"]
tickers.sort()
print(tickers)
````

**`sorted()` leaves the list alone and returns a *new*, sorted list.** Use this when you need a sorted view but want to keep the original order intact.

Run the code cell below and watch the second `print`:

````python
tickers = ["MSFT", "AAPL", "GOOGL", "AMZN"]

print(sorted(tickers))
print(tickers)
````

`sorted(tickers)` gave us an alphabetical copy, but `tickers` itself is still in its original order. This is the same distinction you saw in the last lab between a method that changes a value and one that only changes how it's displayed. Here, `.sort()` changes the list, `sorted()` doesn't.

Both accept **`reverse=True`** to sort from high to low (or Z to A):

````python
daily_prices = [101.2, 100.8, 103.5, 102.1, 104.0]
print(sorted(daily_prices, reverse=True))

tickers = ["MSFT", "AAPL", "GOOGL", "AMZN"]
tickers.sort(reverse=True)
print(tickers)
````

One caution: sorting only works when the items can be compared to each other. A list mixing strings and numbers (`[3, "apple", 1]`) raises a `TypeError` when you try to sort it. And sorting strings is case-sensitive: capital letters come before lowercase ones, so `"Zebra"` sorts *before* `"apple"`.

### Reversing a List: `.reverse()`

If you just want to flip a list end-to-end (not sort it, just reverse the order it's already in), use `.reverse()`. Like `.sort()`, it works **in place** and is permanent.

Run the code cell below:

````python
timeline = ["2019", "2020", "2021", "2022", "2023"]
timeline.reverse()
print(timeline)
````

`.reverse()` does *not* sort; it takes whatever order the list is in and turns it around. If you ever need to undo it, just call `.reverse()` a second time to flip it back.

### How Many Items? `len()`

The `len()` function you used to count characters in a string does the same job for a list: it tells you **how many items** are in it. Same function, same syntax (the list goes inside the parentheses, no dot).

Run the code cell below:

````python
reviews = ["great product", "not worth the price", "will buy again", "disappointed"]
print(len(reviews))

features = ["word_count", "avg_sentence_length", "positive_words"]
print(f"The model uses {len(features)} features.")
````

### Avoiding the `IndexError`

The single most common mistake with lists: asking for an index that doesn't exist.

If a list has 3 items, its valid indexes are `0`, `1`, and `2`. There is no `features[3]` (that's a *fourth* slot, and the list only has three). Ask for it and Python stops with an `IndexError: list index out of range`.

Run the code cell below and read the error:

````python
features = ["word_count", "avg_sentence_length", "positive_words"]

# This list has 3 items, so the only valid indexes are 0, 1, and 2.
print(features[3])
````

The fix comes down to one fact worth memorising:

> **The last valid index of a list is `len(list) - 1`, not `len(list)`.**

A 3-item list has length `3`, so its last index is `2`. Here are the safe ways to reach the end of a list without miscounting:

````python
features = ["word_count", "avg_sentence_length", "positive_words"]

last_index = len(features) - 1
print(features[last_index])   # works it out from the length
print(features[-1])           # simpler: just ask for the last item
````

A few habits that keep `IndexError` away:

* Reach for `list[-1]` instead of counting to the end yourself.
* Remember that an **empty list** (`[]`) has *no* valid indexes at all; even `list[0]` fails on it.
* If an index comes from a calculation or from user input, check the list is long enough (`len(list)`) before you trust it.

### Looking Ahead: Working Through a Whole List

Almost everything useful you'll do with a list involves visiting **every item in turn**: lowercasing every word in a review, adding up every daily return, scanning every headline for a company name.

Python does this with a **`for` loop**, which is the subject of the next lab. Here's the preview so the idea isn't a total surprise:

```python
for feature in features:
    print(feature)
```

Read it almost as English: *for each `feature` in `features`, print it*. The loop hands you one item at a time until the list runs out.

Later, once loops feel natural, you'll meet a compact one-line form called a **list comprehension** (for example, `[f.upper() for f in features]` builds a new list with every feature upper-cased). There's no rush to learn it now; the plain loop above is all you need to follow along until we get there.

### Recap: Your List Toolkit

| Tool | What It Does | Example |
|---|---|---|
| `[ ]` | Create a list | `["AAPL", "MSFT"]` |
| `list[0]` | Access by index (**counting starts at 0**) | `features[0]` → 1st item |
| `list[-1]` | Access from the end | `features[-1]` → last item |
| `list[i] = x` | Replace the item at position `i` | `watchlist[1] = "NVDA"` |
| `.append(x)` | Add `x` to the end | `watchlist.append("AMZN")` |
| `.insert(i, x)` | Add `x` at position `i` | `watchlist.insert(0, "AMZN")` |
| `del list[i]` | Delete the item at position `i` (gone for good) | `del watchlist[0]` |
| `.pop()` / `.pop(i)` | Remove **and return** the last item / item `i` | `x = watchlist.pop()` |
| `.remove(x)` | Remove the first item equal to `x` | `watchlist.remove("NVDA")` |
| `.sort()` | Sort the list in place (**permanent**) | `tickers.sort()` |
| `sorted(list)` | Return a **new** sorted list (original unchanged) | `sorted(tickers)` |
| `reverse=True` | Sort high → low / Z → A | `tickers.sort(reverse=True)` |
| `.reverse()` | Flip the list's order (**permanent**, not a sort) | `timeline.reverse()` |
| `len(list)` | Count the items | `len(reviews)` → `4` |

Keep two things in mind above all: **indexes start at 0**, so the last one is `len(list) - 1`; and **`.sort()` changes the list while `sorted()` doesn't**.

### Tuples: Lists That Can't Change

Sometimes you have a collection of values that belong together and should **never** be edited once set: the pair of numbers describing a screen size, a fixed set of category names, a `(latitude, longitude)` coordinate. Using a list for these works, but nothing stops a stray line of code from quietly changing one of the values.

A **tuple** is Python's answer. It's almost identical to a list, with one deliberate restriction: once you create it, you **cannot** change, add, or remove any of its items. You write one with **parentheses** `( )` instead of square brackets:

```python
sentiment_labels = ("negative", "neutral", "positive")
```

Run the code cell below to create two tuples and check their type:

````python
sentiment_labels = ("negative", "neutral", "positive")
report_shape = (1000, 12)   # 1000 rows, 12 columns

print(sentiment_labels)
print(report_shape)
print(type(sentiment_labels))
````

Python prints a tuple much like a list, but wrapped in parentheses, and `type()` confirms it's a `tuple` rather than a `list`.

### Accessing Elements in a Tuple

Here's the good news: **everything you learned about reading a list works exactly the same on a tuple.** Indexing starts at `0`, negative indexes count from the end, and `len()` counts the items.

Run the code cell below:

````python
sentiment_labels = ("negative", "neutral", "positive")

print(sentiment_labels[0])
print(sentiment_labels[-1])
print(len(sentiment_labels))
````

No surprises: `[0]` is the first item, `[-1]` is the last, `len()` is `3`. The only thing that changes with a tuple is what happens when you try to *write* to it.

### Trying to Change a Tuple

Let's see the restriction in action. Run the code cell below and read the error:

````python
sentiment_labels = ("negative", "neutral", "positive")
sentiment_labels[0] = "very negative"
````

Python stops with `TypeError: 'tuple' object does not support item assignment`. And it's not just assignment: a tuple has **no** `.append()`, no `.insert()`, no `.remove()`, and `del sentiment_labels[0]` fails too. Once written, a tuple is frozen solid.

That's not a limitation to work around; it's the entire reason tuples exist. If some other part of your program tries to modify a value that was never meant to change, Python catches the mistake immediately instead of letting a silent bug slip through.

### Replacing a Whole Tuple

You can't change the *items* inside a tuple, but you can always point the variable at a **brand-new** tuple:

Run the code cell below:

````python
dimensions = (1920, 1080)
print(dimensions)

dimensions = (2560, 1440)
print(dimensions)
````

This isn't modifying the original tuple; it's discarding it and building a fresh one. The distinction matters: `dimensions[0] = 2560` is forbidden, but `dimensions = (2560, 1440)` is completely fine.

### Unpacking a Tuple

A very handy trick: you can assign every value in a tuple to its own variable in a single line, just by listing the variables on the left, separated by commas. This is called **unpacking**.

Run the code cell below:

````python
coordinates = (37.98, 23.73)
latitude, longitude = coordinates

print(latitude)
print(longitude)
````

Python matched the values up in order: `latitude` got the first, `longitude` got the second. The number of variables on the left must match the number of items in the tuple, or you'll get an error.

Unpacking works on lists too, but you'll see it most with tuples, especially later on, when you meet functions that hand back several values at once.

### List or Tuple?

They do almost the same job, so how do you choose?

* Reach for a **list** by default. Most collections you build up, filter, sort, or grow over time should be lists.
* Choose a **tuple** when the collection is a **fixed record** (a set of values that describes one thing and should stay exactly as written). It signals to anyone reading your code: *"these go together and won't change."*

One small gotcha worth knowing: a tuple with a *single* item needs a trailing comma: `("AAPL",)`, not `("AAPL")`. Without the comma, Python just sees parentheses around a normal string.

### Recap: Tuples

| Tool | What It Does | Example |
|---|---|---|
| `( )` | Create a tuple | `("negative", "neutral", "positive")` |
| `tuple[0]` / `tuple[-1]` | Access by index (same rules as lists) | `labels[0]` → 1st item |
| `len(tuple)` | Count the items | `len(labels)` → `3` |
| `a, b = tuple` | Unpack values into separate variables | `lat, lon = coordinates` |
| *(no change allowed)* | `tuple[i] = x`, `.append()`, `.remove()`, `del` all raise errors | n/a |

A tuple is a list you've promised not to change, and Python holds you to the promise.

### Dictionaries: Looking Things Up by Name

Lists and tuples answer the question *"what's at position 2?"*. But very often that's not the question you actually have. You want to know: *"how many times does the word 'revenue' appear?"*, *"what's the price of AAPL?"*, *"what sentiment did document 5 get?"*

A **dictionary** is built for exactly this. Instead of storing values in a numbered sequence, it stores them as **key-value pairs**: a meaningful label (the **key**) paired with the information it points to (the **value**). You look things up by key, not by position.

Think of a paper dictionary: you look up a *word* to find its *definition*. Or a phone contacts list: a *name* points to a *number*.

You write one with **curly braces** `{ }`, each pair as `key: value`, pairs separated by commas:

```python
word_counts = {"revenue": 12, "loss": 3, "growth": 7}
```

Run the code cell below:

````python
word_counts = {"revenue": 12, "loss": 3, "growth": 7}

print(word_counts)
print(type(word_counts))
````

Here the keys are the words (`"revenue"`, `"loss"`, `"growth"`) and the values are how many times each appeared. This word → count mapping is one of the most common structures in all of text analytics.

### Accessing a Value

To get a value out, put its **key** in square brackets (same brackets as a list, but a label goes inside instead of a number).

Run the code cell below:

````python
word_counts = {"revenue": 12, "loss": 3, "growth": 7}

print(word_counts["revenue"])
print(word_counts["growth"])
````

`word_counts["revenue"]` gives `12`. Notice that **position is irrelevant**: it doesn't matter whether `"revenue"` was written first, last, or in the middle. You asked for it by name, and that's what you got back.

### Adding and Changing Entries

Both use the same syntax: assign a value to a key.

* If the key is **new**, Python creates the pair.
* If the key **already exists**, Python overwrites its value.

Run the code cell below:

````python
word_counts = {"revenue": 12, "loss": 3}

word_counts["growth"] = 7      # new key    -> adds the pair
word_counts["loss"] = 5        # existing key -> updates the value
print(word_counts)
````

Dictionaries are **mutable**, like lists: you can reshape them freely after creating them. Starting from an empty dictionary (`{}`) and filling it in as you go is a very common pattern, especially when counting things.

### Removing Entries: `del` and `pop()`

The same two tools you used on lists, adapted to work with keys instead of positions:

* **`del dict[key]`** removes the pair and discards it.
* **`dict.pop(key)`** removes the pair and **returns the value**, so you can keep using it.

Run the code cell below:

````python
stock_prices = {"AAPL": 190.5, "MSFT": 415.2, "GOOGL": 175.8}

del stock_prices["GOOGL"]
print(stock_prices)

removed_price = stock_prices.pop("AAPL")
print(removed_price)
print(stock_prices)
````

Just like with lists, `del` throws the value away while `pop()` hands it back. The only difference here is that you name a **key** rather than a position.

### Avoiding the `KeyError`

Ask a dictionary for a key that doesn't exist, and Python stops with a `KeyError` (the dictionary's version of the `IndexError` you saw with lists).

Run the code cell below and read the error:

````python
stock_prices = {"AAPL": 190.5, "MSFT": 415.2}
print(stock_prices["TSLA"])
````

`KeyError: 'TSLA'` tells you there's no such key. There are two clean ways to avoid this:

1. **Check first with `in`.** On a dictionary, `in` tests the **keys**: `"TSLA" in stock_prices` gives back `True` or `False`.
2. **Use `.get(key)` instead of `[key]`.** If the key is missing, `.get()` returns `None` rather than crashing, and you can supply your own fallback value as a second argument, e.g. `.get("TSLA", 0)`.

Run the code cell below:

````python
stock_prices = {"AAPL": 190.5, "MSFT": 415.2}

print("TSLA" in stock_prices)
print(stock_prices.get("TSLA"))
print(stock_prices.get("TSLA", 0))
````

`.get("TSLA", 0)` is especially handy when counting words: *"give me the current count for this word, or `0` if I haven't seen it yet."*

### Keys, Values, and Length

Run the code cell below:

for word in word_counts:

### Sets: Collections of Unique Values

A **set** is an unordered collection in which every value appears **at most once**. You write one with **curly braces**, like a dictionary, but with plain values instead of `key: value` pairs:

```python
colors = {"navy", "black", "ivory"}
```

Two properties make sets different from everything else in this lab:

* **No duplicates.** Add the same value twice and the set simply keeps one copy.
* **No order.** A set doesn't remember the order you wrote things in, so there is no `[0]` and no slicing.

Run the code cell below and look closely at what gets printed:

````python
sizes = {"M", "S", "M", "L", "S", "M"}

print(sizes)
print(len(sizes))
````

Six values went in, but only three come out: `"M"` and `"S"` were stored once each. The printed order may also look different from what you typed (and may differ between computers). That's normal for a set, and it's why you never ask a set for "the first item".

### Removing Duplicates: `set()`

The most common way to make a set is to pass a list to **`set()`**. This is the quickest way to answer a very common question in text analytics: *how many **different** words does this text use?* (a text's **vocabulary**).

Run the code cell below:

````python
review_words = ["love", "this", "dress", "love", "the", "color", "love", "the", "fit"]

unique_words = set(review_words)

print(unique_words)
print(f"Total words: {len(review_words)}")
print(f"Unique words: {len(unique_words)}")
````

Nine words in the list, but only six different ones. `len(set(...))` is the standard way to measure vocabulary size, and you'll use it again once we work with real reviews.

If you want the unique values back in a predictable order, hand the set to `sorted()` (from earlier in this lab). It returns an ordinary **list**:

````python
print(sorted(unique_words))
````

> **Careful with empty sets.** `{}` creates an empty **dictionary**, not an empty set (dictionaries got the curly braces first). To create an empty set, write `set()`.

````python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
````

### Adding and Removing: `add()`, `remove()`, `discard()`

Sets are mutable, like lists and dictionaries. Because there's no position, the methods are simpler:

* **`.add(x)`** puts `x` in the set (nothing happens if it's already there).
* **`.remove(x)`** takes `x` out, and raises a `KeyError` if it isn't there.
* **`.discard(x)`** takes `x` out if it's there, and quietly does nothing if it isn't.

Run the code cell below:

````python
stopwords = {"the", "a", "and"}

stopwords.add("is")
stopwords.add("the")
print(stopwords)

stopwords.discard("a")
stopwords.discard("xyz")
print(stopwords)
````

Adding `"the"` a second time changed nothing, and discarding `"xyz"` (which was never there) didn't crash. If you used `.remove("xyz")` instead, you'd get the same kind of `KeyError` you saw with dictionaries. Use `.discard()` when you don't care whether the value was there.

### Is It In There? `in` With Sets

`in` works on sets exactly as it does on lists: `"the" in stopwords` gives `True` or `False`. The difference is speed. To answer `in` on a list, Python checks the items one by one; a set jumps straight to the answer, no matter how big it is. With a list of 5 items you'll never notice, but with a vocabulary of 50,000 words checked millions of times, it matters a lot. That's why lists of words to ignore (**stopwords**) are almost always stored as sets.

Run the code cell below:

````python
stopwords = {"the", "a", "and", "is", "this"}

print("the" in stopwords)
print("dress" in stopwords)
````

### Comparing Sets: `&`, `|`, and `-`

Sets come with the operations you may remember from school maths. They're a quick way to compare the words used in two texts.

| Operator | Name | Gives you... |
|---|---|---|
| `a & b` | intersection | values in **both** `a` and `b` |
| `a \| b` | union | values in **either** `a` or `b` (or both) |
| `a - b` | difference | values in `a` but **not** in `b` |

Run the code cell below:

````python
positive_review = {"love", "soft", "fit", "perfect", "color"}
negative_review = {"return", "fit", "small", "color", "cheap"}

print("In both:      ", positive_review & negative_review)
print("In either:    ", positive_review | negative_review)
print("Only positive:", positive_review - negative_review)
print("Only negative:", negative_review - positive_review)
````

`fit` and `color` show up in both reviews, so they don't tell you much about whether a review is positive or negative. Words that appear on only one side (`perfect`, `cheap`, `return`) are much more telling. Keep this idea in mind: later in the course you'll use exactly this comparison on thousands of real reviews.

Notice that `a - b` and `b - a` give different answers. Difference depends on which set comes first, while `&` and `|` don't.

### Recap: Sets

| Tool | What It Does | Example |
|---|---|---|
| `{ }` with values | Create a set | `{"navy", "black"}` |
| `set(list)` | Build a set from a list (duplicates dropped) | `set(["a", "b", "a"])` → `{"a", "b"}` |
| `set()` | Create an **empty** set (`{}` is an empty dict!) | `seen = set()` |
| `.add(x)` | Add a value | `stopwords.add("is")` |
| `.remove(x)` | Remove a value (`KeyError` if missing) | `stopwords.remove("is")` |
| `.discard(x)` | Remove a value if present, no error if missing | `stopwords.discard("is")` |
| `x in set` | Fast membership test | `"the" in stopwords` |
| `a & b` | Values in both | `words_a & words_b` |
| `a \| b` | Values in either | `words_a \| words_b` |
| `a - b` | Values in `a` but not in `b` | `words_a - words_b` |
| `len(set(list))` | Count the distinct values | `len(set(words))` |

### Choosing a Data Structure

| Structure | Written with | Look things up by | Can you change it? | Duplicates? | Reach for it when... |
|---|---|---|---|---|---|
| **List** | `[ ]` | position (`0`, `1`, `2`, …) | yes | allowed | you have an ordered collection you'll grow, sort, or filter |
| **Tuple** | `( )` | position (`0`, `1`, `2`, …) | **no** | allowed | the values are a fixed record that must not change |
| **Dictionary** | `{key: value}` | a meaningful key | yes | keys unique | you want to look values up by name, not by position |
| **Set** | `{ }` / `set()` | membership only (`in`) | yes | **never** | you need unique values, fast `in` checks, or to compare two groups |
